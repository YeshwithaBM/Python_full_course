```python
import string
def encrypt(message_e,shift):
    letter=list(string.ascii_lowercase)
    decrypt_m=[]
    for i in range(len(message_e)):
        if message_e[i]==' ':
            decrypt_m.append(message_e[i])
            continue
        index_letter=letter.index(message_e[i])
        shift_letter=(index_letter+shift)%26
        decrypt_m.append(letter[shift_letter])
    value="".join(decrypt_m)
    return value
def decrypt(message_e,shift):
    letter=list(string.ascii_lowercase)
    encrypt_m=[]
    for i in range(len(message_e)):
        if message_e[i]==' ':
            encrypt_m.append(message_e[i])
            continue
        index_letter=letter.index(message_e[i])
        shift_letter=(index_letter-shift)%26
        encrypt_m.append(letter[shift_letter])
    value="".join(encrypt_m)  
    return value
def caesarCipher():
    while True :
            ask=input("Type 'encrypt' for encryption or 'decrypt' for decryption:")
            message=input("Type your message")
            shift=int(input("type the shift number"))
            if ask=='encrypt':
                print(encrypt(message,shift))
            elif ask=='decrypt':
                print(decrypt(message,shift))
            else:
                print("invalid type, please Type 'encrypt' for encryption or 'decrypt' for decryption: ")
            again_ask=input("Type 'yes' if you want to go again. Otherwise type 'no")
            if again_ask.lower()=='no':
                break
    print("good bye!!!")

caesarCipher()
```

### tutorial solution 

```python
alphabet = [
    'a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j', 'k', 'l', 'm',
    'n', 'o', 'p', 'q', 'r', 's', 't', 'u', 'v', 'w', 'x', 'y', 'z'
]


def encryption(plain_text, shift_key):
    cipher_text = ""

    for char in plain_text:
        if char in alphabet:
            position = alphabet.index(char)
            new_position = (position + shift_key) % 26
            cipher_text += alphabet[new_position]
        else:
            cipher_text += char

    print(f"Here's the text after encryption: {cipher_text}")


def decryption(cipher_text, shift_key):
    plain_text = ""

    for char in cipher_text:
        if char in alphabet:
            position = alphabet.index(char)
            new_position = (position - shift_key) % 26
            plain_text += alphabet[new_position]
        else:
            plain_text += char

    print(f"Here's the text after decryption: {plain_text}")


while True:

    direction = input(
        "Type 'encrypt' for encryption, type 'decrypt' for decryption: "
    ).lower()

    text = input("Type your message: ").lower()

    shift = int(input("Type the shift number: "))

    if direction == "encrypt":
        encryption(text, shift)

    elif direction == "decrypt":
        decryption(text, shift)

    else:
        print("Invalid input.")

    again = input("Type 'yes' if you want to go again. Otherwise type 'no': ").lower()

    if again == "no":
        print("Goodbye!")
        break
```

### Ai solution

```python
import string

alphabet = string.ascii_lowercase


def caesar_cipher(message, shift, direction):
    result = ""

    if direction == "decrypt":
        shift = -shift

    for char in message:
        if char in alphabet:
            position = alphabet.index(char)
            new_position = (position + shift) % 26
            result += alphabet[new_position]
        else:
            result += char

    return result


while True:
    direction = input(
        "Type 'encrypt' for encryption or 'decrypt' for decryption: "
    ).lower()

    message = input("Type your message: ").lower()
    shift = int(input("Type the shift number: "))

    if direction == "encrypt" or direction == "decrypt":
        result = caesar_cipher(message, shift, direction)
        print(f"Result: {result}")
    else:
        print("Invalid direction.")

    again = input(
        "Type 'yes' if you want to go again. Otherwise type 'no': "
    ).lower()

    if again == "no":
        print("Goodbye!")
        break
```
