```python
def findNumber(nums):
    even_digits=0
    for i in range(len(nums)):
        count=len(str(nums[i]))
        if count%2==0:
            even_digits+=1
    return even_digits
nums=[112,345,211,116,7898]
print(findNumber(nums))
```

## LeetCode 1732 — Find the Highest Altitude

```python
def largestAltitude(gain):
        prefix_sum=[0]
        ps_sum=0
        for alt in gain:
            ps_sum+=alt
            prefix_sum.append(ps_sum)
        return max(prefix_sum)
gain=[-5,1,5,0,-7]
print(largestAltitude(gain))
```
