```python
def runningSum( nums):
        sum=0
        for i in range(len(nums)):
            sum+=nums[i]
            nums[i]=sum
        return nums
nums=[1,2,3,4]
print(runningSum(nums))
```

```python
def maximumWealth(accounts):
        max_num=0
        for i in range(len(accounts)):
            sum_num=0
            for j in range(len(accounts[i])):
                sum_num+=accounts[i][j]
            if sum_num>max_num:
                max_num=sum_num
        return max_num
nums=[[1,2,3],[3,4,5],[10,20,30]]
print(maximumWealth(nums))
```

```python
def buildArray(nums):
    ans=[]
    for i in range(len(nums)):
        ans.append(nums[nums[i]])
    return ans
nums=[5,0,1,2,3,4]
print(buildArray(nums))
            
```

```python
def findNumbers(nums):
    even_digits=0
    for i in range(len(nums)):
        count=0
        nums_str=str(nums[i])
        for j in range(len(nums_str)):
            count+=1
        if count%2==0:
            even_digits+=1
    return even_digits
nums=[2,456,23,78,8907,3]
print(findNumbers(nums))
```

```python
def isSubsequence(s,t):
    ans=False
    for i in range(len(s)):
        if s[i] in t:
            ans=True
        else:
            return False
    return ans
s="abc"
t="ahbgdc"
print(isSubsequence(s,t))
```
