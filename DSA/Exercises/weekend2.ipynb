```python
#leetcode 1480
def runningSum( nums):
        sum=0
        for i in range(len(nums)):
            sum+=nums[i]
            nums[i]=sum
        return nums
nums=[1,2,3,4]
print(runningSum(nums))
```

```python
#leetcode 1672
def maximumWealth(accounts):
        max_num=0
        for i in range(len(accounts)):
            sum_num=0
            for j in range(len(accounts[i])):
                sum_num+=accounts[i][j]
            if sum_num>max_num:
                max_num=sum_num
        return max_num
nums=[[1,2,3],[3,4,5],[10,20,30]]
print(maximumWealth(nums))
```

```python
#leetcode 1920
def buildArray(nums):
    ans=[]
    for i in range(len(nums)):
        ans.append(nums[nums[i]])
    return ans
nums=[5,0,1,2,3,4]
print(buildArray(nums))
            
```

```python
#leetcode 1295
def findNumbers(nums):
    even_digits=0
    for i in range(len(nums)):
        count=0
        nums_str=str(nums[i])
        for j in range(len(nums_str)):
            count+=1
        if count%2==0:
            even_digits+=1
    return even_digits
nums=[2,456,23,78,8907,3]
print(findNumbers(nums))
```

```python
#leetcode 88
def merge(nums1,m,nums2,n):
    left=m-1
    right=n-1
    point=m+n-1
    while right>=0:
        if left>=0 and  nums1[left]>nums2[right]:
            nums1[point]=nums1[left]
            left-=1
        else:
            nums1[point]=nums2[right]
            right-=1
        point-=1
    return nums1
nums1=[1,2,3,0,0,0]
nums2=[2,5,6]
print(merge(nums1,3,nums2,3))
```

```python
#leetcode 392
def isSubsequence(s,t):
    if not s:
        return True
    left=0
    right=0
    while right<len(t):
        if s[left]==t[right]:
            left+=1
        right+=1
        if left==len(s):
            return True
    return False

s="axc"
t="ahbgdc"
print(isSubsequence(s,t))
```

```python
#leetcode 852
def peakIndexInMountainArray(arr):
    low=0
    high=len(arr)-1
    while low<high:
        mid=(low+high)//2
        if arr[mid]>arr[mid+1]:
            high=mid
        else:
            low=mid+1
    return low
arr=[0,10,2,1]
print(peakIndexInMountainArray(arr))
```

```python

def majorityElement(nums):
    if len(nums)==1:
        return nums[0]
    count={}
    max_e=0
    max_k=0
    for i in range(len(nums)):
        if nums[i] in count:
            count[nums[i]]+=1
        else:
            count[nums[i]]=1
    for key,value in count.items():
        if value>max_e:
            max_e=value
            max_k=key
        
    return max_k
nums=[2,2,2,2,23,39,8,1,1,1]
print(majorityElement(nums))
```
