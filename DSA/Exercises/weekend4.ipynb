### Medium — LC 739: Daily Temperatures
#### Brute force

```python
from typing import List
class Solution:
    def dailyTemperatures(self, temperatures: List[int]) -> List[int]:
        stack=[]
        answer=[]
        left=0
        stack.append(temperatures[0])
        while left<=len(temperatures)-1:
            for i in range(left+1,len(temperatures)):
                if stack[-1]<temperatures[i]:
                    answer.append(i-left)
                    left+=1
                    stack.append(temperatures[left])
                    break
            else:
                answer.append(0)
                left+=1
                if left==len(temperatures):
                    break
                stack.append(temperatures[left])


        return answer
obj=Solution()
obj.dailyTemperatures([73,74,75,71,69,72,76,73])  
```

```python
from typing import List

class Solution:
    def dailyTemperatures(self, temperatures: List[int]) -> List[int]:
        answer = [0] * len(temperatures)
        stack = []

        for i in range(len(temperatures)):

            while stack and temperatures[i] > temperatures[stack[-1]]:
                previous = stack.pop()
                answer[previous] = i - previous

            stack.append(i)

        return answer
obj=Solution()
obj.dailyTemperatures([73,74,75,71,69,72,76,73])  
```
