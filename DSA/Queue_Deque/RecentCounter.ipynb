### Number of Recent Calls — LC 933

### Brute force


| Approach              |       Per `ping()` | Overall for n pings | Space |
| --------------------- | -----------------: | ------------------: | ----: |
| Brute Force           |               O(n) |           **O(n²)** |  O(n) |
| Queue + Front Pointer | Amortized **O(1)** |            **O(n)** |  O(n) |


```python
class RecentCounter:

    def __init__(self):
        self.times=[]
        self.count=0

    def ping(self, t: int) -> int:
        self.times.append(t)
        self.start=t-3000
        self.count=0
        for i in range(len(self.times)):
            if self.start<=self.times[i]<=t:
                self.count+=1
        return self.count


obj = RecentCounter()
param_1 = obj.ping(1)
param_2 = obj.ping(100)
param_3 = obj.ping(3001)
param_4 = obj.ping(3002)
lst=[param_1,param_2,param_3,param_4]
print(lst)
```

## Optimized

```python
class RecentCounter:

    def __init__(self):
        self.times=[]
        self.count=0
        self.front=0
    def ping(self, t: int) -> int:
        self.times.append(t)
        self.start=t-3000
        while self.times[self.front]<self.start:
            self.front+=1
 
        self.count=len(self.times)-self.front

        return self.count



obj = RecentCounter()
param_1 = obj.ping(1)
param_2 = obj.ping(100)
param_3 = obj.ping(3001)
param_4 = obj.ping(3002)
lst=[param_1,param_2,param_3,param_4]
print(lst)
```
