### Deque — Double-Ended Queue

A Deque (pronounced “deck”) is a queue where we can insert and remove from both ends.
- usually in queue we add at end and we remove at first which follows FIFO
- but in deque either add at first(front)  or at end(rear) and either remove at first(front)  or at end(rear) 

```python
class Deque:

    def __init__(self):
        self.items = []

    def add_front(self, value):
        self.items.insert(0, value)

    def add_rear(self, value):
        self.items.append(value)

    def remove_front(self):
        return self.items.pop(0)

    def remove_rear(self):
        return self.items.pop()

    def front(self):
        return self.items[0]

    def rear(self):
        return self.items[-1]

    def is_empty(self):
        return len(self.items) == 0
d = Deque()

d.add_rear(10)
d.add_rear(20)
d.add_front(5)

print(d.items)
d.remove_front()
d.remove_rear()
print(d.items)
```

### Monotonic Deque
- "Find the maximum/minimum value in every sliding window."
- LC 239
#### brute force 

```python
from typing import List
class Solution:
    def maxSlidingWindow(self, nums: List[int], k: int) -> List[int]:
        ans=[]
        left=k
        right=0
        while left<=len(nums):
            max_elemnt=nums[right]
            for x in range(right+1,left):
                if nums[x]>max_elemnt:
                   max_elemnt=nums[x]
            right+=1
            left+=1
            ans.append(max_elemnt)
        return ans
obj=Solution()
obj.maxSlidingWindow([1,3,-1,-3,5,3,6,7],3)
```

#### Optimized

```python
from typing import List
from collections import deque


class Solution:

    def maxSlidingWindow(self, nums: List[int], k: int) -> List[int]:

        deq = deque()   # stores indices
        ans = []

        for i in range(len(nums)):

            # 1. Remove smaller elements from the back
            while deq and nums[deq[-1]] < nums[i]:
                deq.pop()

            # 2. Add current index
            deq.append(i)

            # 3. Remove elements outside the current window
            if deq[0] <= i - k:
                deq.popleft()

            # 4. Once we have a complete window,
            #    front of deque contains the maximum
            if i >= k - 1:
                ans.append(nums[deq[0]])

        return ans


obj = Solution()

print(obj.maxSlidingWindow(
    [1, 3, -1, -3, 5, 3, 6, 7],
    3
))
```

```python
from typing import List
from collections import deque

class Solution:
    def maxSlidingWindow(self, nums: List[int], k: int) -> List[int]:
        q = deque()
        ans = []

        # Window is from [i - k + 1 : i]
        for i, x in enumerate(nums):

            # Remove indices to the left of the window from the queue
            while q and q[0] <= i - k:
                q.popleft()

            # Maintain decreasing queue
            while q and nums[q[-1]] < x:
                q.pop()
            q.append(i)

            # When window is formed, start appending:
            if i - k + 1 >= 0:
                ans.append(nums[q[0]])

        return ans
obj = Solution()

print(obj.maxSlidingWindow(
    [1, 3, -1, -3, 5, 3, 6, 7],
    3
))
```
