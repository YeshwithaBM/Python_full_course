### Queue 
- FIFO First in first out.
Implement Queue using Python list

We want these operations:

- Enqueue → add at the rear
- Dequeue → remove from the front
- Front → see the first element
- is_empty → check whether queue is empty

Our implementation works, but dequeue() is O(n):

self.queue.pop(0)

because Python has to shift all remaining elements.

-That's actually useful for us because this is exactly the inefficiency we'll optimize later.

```python
class Queue:

    def __init__(self):
        self.queue = []

    def enqueue(self, x):
        self.queue.append(x)

    def dequeue(self):
        if len(self.queue) == 0:
            return None
        return self.queue.pop(0)

    def front(self):
        if len(self.queue) == 0:
            return None
        return self.queue[0]

    def is_empty(self):
        return len(self.queue) == 0
```

```python
q = Queue()

q.enqueue(10)
q.enqueue(20)
q.enqueue(30)

print(q.queue)       # [10, 20, 30]

print(q.dequeue())   # 10
print(q.front())     # 20
print(q.queue)       # [20, 30]
```

- Now let's understand why pop(0) is slow and how we can fix it
- Optimization: use a front pointer
- Instead of physically deleting the first element, we can simply remember where the front currently is.
- Instead of shifting/deleting elements, move a pointer/index that represents the front.

```python
q=[]
q.append(10)
q.append(20)
q.append(30)
print(q)
front=0
# dequeue
x=q[front]
front+=1
print(q[front:])
```

```python
class Queue:
    def __init__(self):
        self.q = []
        self.front_index = 0

    def enqueue(self, value):
        self.q.append(value)

    def front(self):
        return self.q[self.front_index]

    def dequeue(self):
        d = self.q[self.front_index]
        self.front_index += 1
        return d
```

```python
q = Queue()

q.enqueue(10)
q.enqueue(20)
q.enqueue(30)
print(q.front())
print(q.dequeue())
print(q.front())
print(q.dequeue())
print(q.front())
```
