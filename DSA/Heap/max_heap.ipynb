### Max heap Insertion

```python
class MaxHeap:
    def __init__(self):
        self.data=[]

    def parent(self,i):
        return (i-1)//2
    
    def swap(self,i,j):
        self.data[i],self.data[j]= self.data[j],self.data[i]

    def insert(self, value):
        self.data.append(value)
        i = len(self.data) - 1

        while i > 0:
            p_i = self.parent(i)

            if self.data[i] <= self.data[p_i]:
                break

            self.swap(i, p_i)
            i = p_i
```

```python
heap = MaxHeap()

heap.insert(2)
heap.insert(5)
heap.insert(4)
heap.insert(9)
heap.insert(7)
heap.insert(3)

print(heap.data)
```

```python
class MaxHeap:
    def __init__(self):
        self.data=[]

    def parent(self,i):
        return (i-1)//2
    
    def swap(self,i,j):
        self.data[i],self.data[j]= self.data[j],self.data[i]

    def insert(self, value):
        self.data.append(value)
        i = len(self.data) - 1

        while i > 0:
            p_i = self.parent(i)

            if self.data[i] <= self.data[p_i]:
                break

            self.swap(i, p_i)
            i = p_i
    
    def downheap(self, i):
        n = len(self.data)

        while True:
            left = 2 * i + 1
            right = 2 * i + 2
            largest = i

            # Check whether the left child is smaller
            if left < n and self.data[left] > self.data[smallest]:
                largest = left

            # Check whether the right child is smaller
            if right < n and self.data[right] > self.data[smallest]:
                largest = right

            # Heap property is satisfied
            if largest == i:
                break

            # Swap with the smallest child
            self.swap(i, largest)
            i = largest

    def remove_max(self):
        if not self.data:
            raise IndexError("Heap is empty")

        root = self.data[0]

        # Replace root with the last element
        self.data[0] = self.data[-1]
        self.data.pop()

        # Restore heap property
        if self.data:
            self.downheap(0)

        return root
```

```python
heap = MaxHeap()

heap.insert(2)
heap.insert(5)
heap.insert(4)
heap.insert(9)
heap.insert(7)
heap.insert(3)


print(heap.remove_max())
print(heap.data)
```
