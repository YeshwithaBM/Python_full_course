## heapify 

```python
class MinHeap:
    def __init__(self):
        self.data=[]

    def parent(self,i):
        return (i-1)//2
    
    def swap(self,i,j):
        self.data[i],self.data[j]= self.data[j],self.data[i]

    def downheap(self, i):
            n = len(self.data)
    
            while True:
                left = 2 * i + 1
                right = 2 * i + 2
                smallest = i
    
                # Check whether the left child is smaller
                if left < n and self.data[left] < self.data[smallest]:
                    smallest = left
    
                # Check whether the right child is smaller
                if right < n and self.data[right] < self.data[smallest]:
                    smallest = right
    
                # Heap property is satisfied
                if smallest == i:
                    break
    
                # Swap with the smallest child
                self.swap(i, smallest)
                i = smallest

    def heapify(self, arr):
        self.data = list(arr)
        n = len(self.data)

        start = n // 2 - 1

        for i in range(start, -1, -1):
            self.downheap(i)        
        
```

```python
heap = MinHeap()
arr = [9, 4, 7, 1, 3, 6, 2]
heap.heapify(arr)

print(heap.data)
```
