### LC 110 Balanced or not tree

```python
# Creating an unbalanced tree
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right


root = TreeNode(3)

root.left = TreeNode(9)

root.right = TreeNode(20)
root.right.left = TreeNode(15)
root.right.right = TreeNode(7)

# Make the right side deeper
root.right.right.right = TreeNode(8)
root.right.right.right.right = TreeNode(10)
```

```python
class Solution:
    def valid(self, root):
        def height(node):
            if node is None:
                return 0

            left = height(node.left)
            right = height(node.right)
            return 1 + max(left, right)

        left=height(root.left)
        right=height(root.right)
        return abs(left-right)<=1

s=Solution()
s.valid(root)
```

```python
class Solution:
    def valid(self, root):

        def height(node):
            if node is None:
                return 0

            left = height(node.left)

            # Left subtree is already unbalanced
            if left == -1:
                return -1

            right = height(node.right)

            # Right subtree is already unbalanced
            if right == -1:
                return -1

            # Current node is unbalanced
            if abs(left - right) > 1:
                return -1

            # Return height to parent
            return 1 + max(left, right)

        return height(root) != -1


s = Solution()
print(s.valid(root))
```
