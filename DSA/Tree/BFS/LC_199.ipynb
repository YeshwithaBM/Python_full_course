### LC 199 Binary right side view

```python
#creating Tree
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
root = TreeNode(1)
root.left = TreeNode(2)
root.right = TreeNode(3)

root.left.right = TreeNode(5)
root.right.right = TreeNode(4)
```

```python
from collections import deque
def RightView(root):
    if root is None:
        return []
    q=deque([root])
    res=[]
    while q:
        level=[]
        for i in range(len(q)):
            n=q.popleft()
            level.append(n.val)
            if n.left:
                q.append(n.left)
            if n.right:
                q.append(n.right)       
        right=level[-1]
        res.append(right)
    return res  
RightView(root)
```
