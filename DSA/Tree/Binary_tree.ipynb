- A binary tree is where each node has at most two childrens .
1. left 
2. right 
```
        1
       / \
      2   3
     / \
    4   5
```
- 1 → root
- 2, 3 → children of 1
- 4, 5 → children of 2
- 2 → left child of 1
- 3 → right child of 1

### How we represent it in coding problems

````python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

root = TreeNode(1)



root.left = TreeNode(2)
root.right = TreeNode(3)

root.left.left = TreeNode(4)
root.left.right = TreeNode(5)
````
