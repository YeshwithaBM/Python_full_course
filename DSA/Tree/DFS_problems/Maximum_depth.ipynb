#### maximum depth of tree LC 104

```python
#creating Tree
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
root = TreeNode(3)
root.left = TreeNode(9)
root.right = TreeNode(20)

root.right.left = TreeNode(15)
root.right.right = TreeNode(7)
```

```python
def MaxDepth(root):
    if root is None:
        return 0
    l=MaxDepth(root.left)
    r=MaxDepth(root.right)
    return 1+max(l,r)
print(MaxDepth(root))
```
