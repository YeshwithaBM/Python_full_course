### Recursion

Recursion is a programming technique in which a **function calls itself** to solve a problem by breaking it into **smaller or simpler versions of the same problem**.

Instead of solving the entire problem at once, recursion solves a small part and lets another recursive call solve the remaining smaller part.

A recursive function must eventually stop, so every recursive solution needs a **base case**.

---

### 1. Basic Structure of Recursion

A recursive function generally has two important parts:

1. **Base Case** → stops the recursion.
2. **Recursive Case** → function calls itself with a smaller/simpler input.

```python
def function(n):

    # Base case
    if condition:
        return

    # Recursive case
    function(smaller_input)

## Call Stack

The **call stack** is a stack used by the program to keep track of active function calls.

It follows the **LIFO (Last In, First Out)** principle, just like the Stack data structure.

When a function calls another function, the new function call is **pushed onto the call stack**.

When a function finishes, its stack frame is **removed (popped)** from the call stack.

### Example

```python
def count_down(n):
    if n == 0:
        return

    print(n)
    count_down(n - 1)

count_down(3)

#### Key Takeaways
- Recursion = a function calling itself.
- Every recursive solution needs a base case.
- The recursive call should move toward the base case.
- Recursive calls are stored in the call stack.
- The call stack follows LIFO (Last In, First Out).
- The last recursive call is the first one to return.
- Recursion has a going-down phase and an unwinding/returning phase.
- Code before the recursive call executes while going down.
- Code after the recursive call executes while unwinding.
- Recursive calls consume call-stack space.
- Recursion is especially useful for trees, DFS, backtracking, divide and - conquer, and some DP problems.
- Many recursive algorithms can be converted to iterative solutions using an explicit stack.

````python
def count_down(n):
    if n==0: #Base case
        return
    print(n)
    count_down(n-1) #recursive call
count_down(4)
````

````python
def factorial(n):

    if n == 0:
        return 1
    return n * factorial(n - 1)
    
factorial(4)
````

````python
def sum(n):
    if n==0:
        return 0
    return n+sum(n-1)
sum(5)
````

````python

from collections import deque

def sumarray(nums):
    if len(nums)==1:
        return nums[0]
    num=deque(nums)
    num_0=num.popleft()
    return num_0+sumarray(num)
sumarray([2,4,6,8,10])
````

````python
def sumarray(nums, i):

    if i == len(nums):
        return 0

    return nums[i] + sumarray(nums, i + 1)
sumarray([2,4,6,8,10],0)
````

### LC- 509
#### Brute force 
- cause we ae reapeatedly calculating same value 
- Time  = O(2ⁿ)
- Space = O(n)

````python
def fib(n):
    if n==0:
        return 0
    if n==1:
        return 1
        
    return fib(n-1)+fib(n-2)
fib(6)
````

````python
def fib(n):
    memo={0:0,1:1}
    if n in memo:
            return memo[n]
    for i in range(2,n+1):
        memo[i]=fib(i-1)+fib(i-2)
    return memo[n]

fib(7)
````

### optimized
- time is o(n)
- space is o(n)
- Memoization = store previously calculated results and reuse them.

````python
def fib(n, memo):
    if n in memo:
        return memo[n]
    memo[n] = fib(n - 1, memo) + fib(n - 2, memo)
    return memo[n]

memo = {0: 0, 1: 1}

print(fib(700, memo))
````
