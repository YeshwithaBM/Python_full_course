### left sum vs right sum

```python
def pivotIndex(nums):
        left_sum=0
        total=sum(nums)
        for i in range(len(nums)):
            right_sum = total - left_sum - nums[i]
            if right_sum==left_sum:
                return i
            else:
                left_sum+=nums[i]
        return -1
nums=[1,7,3,6,5,6]
print(pivotIndex(nums))  
```

### Range sum query

```python
def runningSum(nums):
    ps=[]
    sum_ps=0
    for i in nums:
        sum_ps+=i
        ps.append(sum_ps)
    return ps
nums=[1,2,3,4]
print(runningSum(nums))
```

```python
from typing import List

class NumArray:

    def __init__(self, nums: List[int]):
        self.ps = []

        sum_ps = 0
        for num in nums:
            sum_ps += num
            self.ps.append(sum_ps)

    def sumRange(self, left: int, right: int) -> int:

        if left == 0:
            return self.ps[right]

        return self.ps[right] - self.ps[left - 1]


nums = [-2, 0, 3, -5, 2, -1]

pr = NumArray(nums)

print(pr.sumRange(0, 2))  # 1
print(pr.sumRange(2, 5))  # -1
print(pr.sumRange(0, 5))  # -3
```
