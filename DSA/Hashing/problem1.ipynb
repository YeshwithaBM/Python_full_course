### 1 — Contains Duplicate
- Given an integer array nums, return True if any value appears at least twice in the array, and return False if every element is distinct.

- Examples:
Input:  [1, 2, 3, 1]
Output: True

```python
#O(n²) worst case. time , for loop and if condition looking all elements 
#O(n) space ,storing in extra ds 
# list membership lookup is O(n)
def containDuplicate(nums):
    present=[]
    for i in nums:
        if i  in present:
            return True
        else:
            present.append(i)
    return False


nums = [1,2,3,1]
containDuplicate(nums)
```

```python
#O(n²) worst case. two for loops 
#O(1) space , i and j
def containDuplicate(nums):
    for i in range(len(nums)):
        for j in range(i+1,len(nums)):
            if nums[i]==nums[j]:
                return True
    return False


nums = [1,2,3,4]
containDuplicate(nums)
```

### Optimal solution

```python
# O(n) for loop and membership lookup is O(1) = O(n x 1) = O(n)
# space is O(n) because of ds to store 
def containDuplicate(nums):
    present=set()
    for i in nums:
        if i in present:
            return True
        else:
            present.add(i)
    return False


nums = [1,2,3,1]
containDuplicate(nums)
```
