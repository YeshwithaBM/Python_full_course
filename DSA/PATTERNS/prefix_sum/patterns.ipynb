### left sum vs right sum

```python
def pivotIndex(nums):
        left_sum=0
        total=sum(nums)
        for i in range(len(nums)):
            right_sum = total - left_sum - nums[i]
            if right_sum==left_sum:
                return i
            else:
                left_sum+=nums[i]
        return -1
nums=[1,7,3,6,5,6]
print(pivotIndex(nums))  
```

### Range sum query

```python
def runningSum(nums):
    ps=[]
    sum_ps=0
    for i in nums:
        sum_ps+=i
        ps.append(sum_ps)
    return ps
nums=[1,2,3,4]
print(runningSum(nums))
```

```python
from typing import List

class NumArray:

    def __init__(self, nums: List[int]):
        self.ps = []

        sum_ps = 0
        for num in nums:
            sum_ps += num
            self.ps.append(sum_ps)
        print(self.ps)

    def sumRange(self, left: int, right: int) -> int:

        if left == 0:
            return self.ps[right]

        return self.ps[right] - self.ps[left - 1]


nums = [1,2,3]

pr = NumArray(nums)

print(pr.sumRange(0, 0))  # 1
print(pr.sumRange(0,1))  # -1
print(pr.sumRange(0, 2))  # -3
print(pr.sumRange(1, 1))  # 1
print(pr.sumRange(1,2))  # -1
print(pr.sumRange(2, 2))  # -3
```

#### prefix sum + Hashmap
#### subarray sum equals k

```python
#O(n^3)
def subarraySum(nums,k):
    subarray=0
    for i in range(len(nums)):
        for j in range(i,len(nums)):
            sum_subarray=sum(nums[i:j+1])
            if sum_subarray==k:
                subarray+=1
    return subarray
nums=[1,2,3]
k=3
print(subarraySum(nums,k))
```

```python
#O(n^2)
def subarraySum(nums,k):
    count=0
    curr_sum = 0
    for i in range(len(nums)):
        curr_sum = 0
        for j in range(i, len(nums)):
            curr_sum += nums[j]
            if curr_sum == k:
                count += 1
    return count
nums=[1,2,3]
k=3
print(subarraySum(nums,k))
```
