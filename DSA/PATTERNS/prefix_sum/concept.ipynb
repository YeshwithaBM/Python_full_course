```python
from typing import List

class NumArray:

    def __init__(self, nums: List[int]):
        self.ps = []

        sum_ps = 0
        for num in nums:
            sum_ps += num
            self.ps.append(sum_ps)

    def sumRange(self, left: int, right: int) -> int:

        if left == 0:
            return self.ps[right]

        return self.ps[right] - self.ps[left - 1]


nums = [-2, 0, 3, -5, 2, -1]

pr = NumArray(nums)

print(pr.sumRange(0, 2))  # 1
print(pr.sumRange(2, 5))  # -1
print(pr.sumRange(0, 5))  # -3
```
