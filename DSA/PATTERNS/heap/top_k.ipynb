### Top K Elements
- Pattern 1: Top K
- Find the K largest, smallest or most frequent elements.

```python
class MinHeap:
    def __init__(self):
        self.data = []

    def parent(self, i):
        return (i - 1) // 2

    def swap(self, i, j):
        self.data[i], self.data[j] = self.data[j], self.data[i]

    def insert(self, value):
        self.data.append(value)
        i = len(self.data) - 1

        # Up-heap
        while i > 0:
            p_i = self.parent(i)

            if self.data[i] >= self.data[p_i]:
                break

            self.swap(i, p_i)
            i = p_i

    def downheap(self, i):
        n = len(self.data)

        while True:
            left = 2 * i + 1
            right = 2 * i + 2
            smallest = i

            # Check whether the left child is smaller
            if left < n and self.data[left] < self.data[smallest]:
                smallest = left

            # Check whether the right child is smaller
            if right < n and self.data[right] < self.data[smallest]:
                smallest = right

            # Heap property is satisfied
            if smallest == i:
                break

            # Swap with the smallest child
            self.swap(i, smallest)
            i = smallest

    def remove_min(self):
        if not self.data:
            raise IndexError("Heap is empty")

        root = self.data[0]

        # Replace root with the last element
        self.data[0] = self.data[-1]
        self.data.pop()

        # Restore heap property
        if self.data:
            self.downheap(0)

        return root
```

```python
def top_k_largest(nums, k):
    if k <= 0:
        return []

    heap = MinHeap()

    for num in nums:

        if len(heap.data) < k:
            heap.insert(num)

        elif num > heap.data[0]:
            heap.remove_min()
            heap.insert(num)

    return heap.data
```

```python
nums = [8, 1, 7, 3, 10, 5]

print(top_k_largest(nums, 3))
```
