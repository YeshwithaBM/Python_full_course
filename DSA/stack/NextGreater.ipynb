### Brute force

```python
def nextGreaterElement(nums1,nums2):
    res=[]
    for left in range(len(nums1)):
        for right in range(len(nums2)):
            if nums1[left]==nums2[right]:
                for j in range(right+1,len(nums2)):
                    if nums2[right]<nums2[j]:
                        res.append(nums2[j])
                        break
                else:
                    res.append(-1)
                break
    return res
nums1=[4,1,3]
nums2=[1,3,4,2]
print(nextGreaterElement(nums1,nums2))
```
