### Brute force

```python
# time= O(n^3)
# space = o(n)
def nextGreaterElement(nums1,nums2):
    res=[]  #o(n)
    for left in range(len(nums1)):#O(n)
        for right in range(len(nums2)):#o(n)
            if nums1[left]==nums2[right]: #o(1)
                for j in range(right+1,len(nums2)):#o(n)
                    if nums2[right]<nums2[j]: #o(1)
                        res.append(nums2[j])#o(1)
                        break
                else:
                    res.append(-1)#o(1)
                break
    return res 
nums1=[4,1,3]
nums2=[1,3,4,2]
print(nextGreaterElement(nums1,nums2))
```

### Optimized 

```python
# Time 
#  for + while loop= O(n)+o(n)
# while loop= O(n)
# for loop = O(m)
# Total=O(n)+o(n)+o(m) = O(n+m)
#space = o(n) + o(n) + o(n)= o(n)

def nextGreaterElement(nums1,nums2):
    stack=[] #o(n)
    result={} #o(n)
    answer=[] #o(n)
    stack.append(nums2[0]) #o(1)
    for i in range(len(nums2)-1): #o(n)
        if stack[-1]>nums2[i+1]: #o(1)
            stack.append(nums2[i+1]) #o(1)
        else:
            while stack and stack[-1]<nums2[i+1]: # o(n)
                result[stack[-1]]=nums2[i+1] #o(1)
                stack.pop() #o(1)
            stack.append(nums2[i+1]) #o(1)
    while stack: #o(n)
        result[stack.pop()] = -1
    for i in nums1: #o(m)
        x=result[i]
        answer.append(x)
    return answer
nums1=[4,1,3]
nums2=[1,3,4,2]
print(nextGreaterElement(nums1,nums2))
```
