#### LeetCode 1047 — Remove All Adjacent Duplicates In String.

```python
#O(n^2) time 
#o(n) space 
def removeadj(s):
    s_l = list(s)
    j = 0
    while j < len(s_l) - 1:
        if s_l[j] == s_l[j + 1]:
            s_l.pop(j)
            s_l.pop(j)
            j = max(0, j - 1)
        else:
            j += 1

    return "".join(s_l)

s = 'abbaca'
print(removeadj(s))
```

### Optimized

```python
#O(n)
#O(n)
def removeadj(s):
    stack=[]
    for i in range(len(s)):
        if stack and s[i] == stack[-1]:
            stack.pop()
        else:
            stack.append(s[i])
    return "".join(stack)

s='abbaca'
print(removeadj(s))
```
