```python

def isValid(s):
    stack=[]
    for i in s:
        if i=='(' or i=='{' or i=='[':
            stack.append(i)
        elif len(stack)!=0:
            x=stack[-1]
            if (x=='(' and i==')') or (x=='{' and i=='}') or (x=='[' and i==']'):
                stack.pop()
            else:
                return False 
        else:
            return False 
    
    if len(stack)!=0:
        return False
    return True
s="(){}[]"
print(isValid(s)) 
    
```
