#### Fast and slow pointer

```python
class Node:
    def __init__(self,data):
        self.data=data
        self.next=None
#create the linked list
first=Node(10)
second=Node(20)
third=Node(30)
first.next=second
second.next=third
```

#### Brute force
- First Traversal  → O(n)
- Second Traversal → O(n)

Total = O(2n)

Big-O ignores constants

= O(n)
- space O(1)

```python
def middleNode(head):
    temp=head
    count=0
    while temp is not None:
        temp=temp.next
        count+=1
    mid=count//2
    slow=head
    for i in range(mid):
        slow=slow.next
    return slow
s=middleNode(first)
print(s.data)
```

### Optimized
- Time is o(n)
- space is o(1)

```python
def middleNode(head):
    slow=head
    fast=head
    while fast and fast.next:
        slow=slow.next
        fast=fast.next.next
    return slow
s=middleNode(first)
print(s.data)
```

#### Leetcode 141 linked list cycle 

#### Brute force 


```python
class Node:
    def __init__(self,data):
        self.data=data
        self.next=None
#create the linked list
first=Node(10)
second=Node(20)
third=Node(30)
first.next=second
second.next=third
third.next=second
def middleNode(head):
    visit=set()
    current=head
    while current is not None:
        if current in visit:
            return True
        visit.add(current)
        current=current.next
    return False
s=middleNode(first)
print(s)
```

### Optimized

```python
def hasCycle(head):
    if head is None:
        return False
    slow=head
    fast=head
    while fast and fast.next:
        slow=slow.next
        fast=fast.next.next
        if slow == fast:
            return True
    return False
s=hasCycle(first)
print(s)
```
