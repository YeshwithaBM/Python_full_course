#### O(1)       → constant
#### O(log n)   → very slow growth
#### O(n)       → linear
#### O(n log n) → slightly faster
#### O(n²)      → quadratic
#### O(2ⁿ)      → exponential
- Big-O describes how the algorithm scales as the input size grows, not how fast it is for one particular input.

### Time Complexity — Best → Worst

| Complexity | Name | Example |
|---|---|---|
| **O(1)** | Constant | Array access |
| **O(log n)** | Logarithmic | Binary Search |
| **O(√n)** | Square-root | Some number algorithms |
| **O(n)** | Linear | Single loop |
| **O(n log n)** | Linearithmic | Merge Sort |
| **O(n²)** | Quadratic | Nested loops |
| **O(n³)** | Cubic | 3 nested loops |
| **O(2ⁿ)** | Exponential | Brute-force Fibonacci |
| **O(n!)** | Factorial | Brute-force permutations |

---

### Space Complexity

The same idea applies to space complexity, but instead of measuring **time**, we measure the **extra memory used by an algorithm**.

| Complexity | Example |
|---|---|
| **O(1)** | A few variables |
| **O(log n)** | Some recursive algorithms |
| **O(n)** | Array/hash table containing `n` elements |
| **O(n log n)** | Some algorithms' auxiliary storage |
| **O(n²)** | 2D matrix |
| **O(2ⁿ)** | Storing all subsets |
| **O(n!)** | Storing all permutations |
