```python
class bankAccount:
    def __init__(self,Ac_holder_name,balance=0):
        self.Ac_holder_name=Ac_holder_name
        self.balance=balance
    def __str__(self):
        return f'Account holder name : {self.Ac_holder_name}\nBalance: {self.balance}'
    def deposit(self,money):
        self.balance+=money
        print(f"Successfully deposited {money} and avl bal is {self.balance}")
    def withdraw(self,money):
        if money>self.balance:
            print("Insufficient Balance")
        else:
            self.balance-=money
            print(f"{money} is debited. avl bal is {self.balance}")
bank=bankAccount("Yeshu",500)
print(bank)
bank.deposit(300)
bank.withdraw(700)
```
