```python
def processStr( s: str) -> str:
        res=[]
        for i,val in enumerate(s):
            if val.isalpha():
                res.append(val)
            elif val=='#':
                res.extend(res)
            elif val=='%':
                res.reverse()
            elif val=="*":
                res.pop()
        ans="".join(res)
        return ans


s = "abc#"
print(processStr(s))
```

```python
# return no of days in a month 
def no_of_days(year,month):
    if (year%4==0 and year%100!=0) or (year%400==0):
        print(f'{year} is leap year.')
        if month>12 or month<1:
            return "invalid month"
        elif month==1 or month==3 or month==5 or month==7 or month==8 or month==10 or month==31:
            return f'number of days in month {month} is 31'
        elif month==2:
            return f'number of days in month {month} is 29'
        else:
            return f'number of days in month {month} is 30'
    else:
        print(f'{year} is not a leap year.')
        if month>12 and month<1:
            return "invalid month"
        elif month==1 or month==3 or month==5 or month==7 or month==8 or month==10 or month==31:
            return f'number of days in month {month} is 31'
        elif month==2:
            return f'number of days in month {month} is 28'
        else:
            return f'number of days in month {month} is 30'
print(no_of_days(2006,2))
    
```

```python
def leap_year(year):
    if year % 4 == 0:
        if year % 100 == 0:
            if year % 400 == 0:
                return True
            else:
                return False
        else:
            return True
    else:
        return False
def days_in_month(year,month):
    days_list=[31,28,31,30,31,30,31,31,30,31,30,31]
    if leap_year(year) and month==2:
        return 29
    else:
        return days_list[month-1]
print(days_in_month(2023,2))
```
