### Functions 
- functions is a block of code where it does specific task when it is called .
- A **function** is a reusable block of code that performs a specific task.  
It is defined once and can be called multiple times, making programs cleaner and reducing code duplication.
##### Defining function
```python
def function_name(parameters):
    # Function body
    return statement
```
> **Note:** Parameters and the `return` statement are **optional**. A function can be created with or without them depending on the requirement.
##### Calling a Function
- A function is called by writing its **name followed by parentheses `()`**. When called, Python executes the code inside the function. If parameters are specified while defining function we need to pass parameters (we call arguements) while calling the function.

```python
function_name(arguements)
```
##### Types of Functions
1. Built-in Functions
Functions that are already provided by Python and can be used directly.
**Examples:**
```python
print("Hello")
len("Python")
input("Enter your name: ")
```
2. User-defined Functions
Functions created by the programmer using the `def` keyword to perform specific tasks.
**Example:**

```python
def greet():
    print("Hello!")

greet()
```

````python
def greet(): #defining the function
    print("hi")
    print("yeshwitha")

greet()#calling the function
greet()#calling the function
greet()#calling the function
````

#### Parameters
- **Parameters** are variables defined inside the parentheses `()` of a function definition.
- They act as placeholders to receive values when the function is called.
- A function can have zero, one, or multiple parameters.
#### Arguments
- **Arguments** are the actual values passed to a function when it is called.
- The values are assigned to the corresponding parameters.
- The number of arguments should match the required parameters (unless default or variable-length parameters are used, which you'll learn later).
> **Tip:**  
> **Parameters** are like variables that receive data.  
> **Arguments** are the actual values passed to those variables.

````python
def greet(name):
    print("hi",name)
greet() # error because greet function require 1 arguement to pass since we specified parameter inside defined function
````

#### Functions with `return`

- `return` sends a value from inside a function back to the place where the function was called.
- The returned value can be stored in a variable, printed, or used in another operation.
- `return` is written inside the function and immediately stops the function's execution.
- If a function has no `return` statement, Python automatically returns `None`.
- **Explicit return:** We explicitly write `return` to send a value back from a function.
- **Implicit return:** When no `return` is written, Python automatically returns `None`.

````python
def add(a,b):
    c= a+b
    print(a+b)
op=add(2,3)
print(op)
````

````python
def add():
    return 'hi'

op=add()
print(op)
````

````python
def add(a,b):
    c= a+b
    return
op=add(2,3)
print(op)
````

### For return, there are explicit and implicit returns.

````python
#Explicit return - You write return yourself:
def add(a, b):
    return a + b

result = add(2, 3)
print(result)   # 5
````

````python
#Implicit return - You don't write return. 
# Python automatically returns None when the function finishes:
def add(a, b):
    print(a + b)

result = add(2, 3)
print(result)
````

### Multiple Return Values

- Multiple values can be returned from a function by separating them with commas; Python packs them into a **tuple** by default.
- If we explicitly return a **list**, the function returns a list.
- If we explicitly return a **set**, the function returns a set.
- If we explicitly return a **dictionary**, the function returns a dictionary.
- The returned collection can then be unpacked into multiple variables when appropriate.

````python
import statistics
def mean_median_mode(list1):
    return [
        statistics.mean(list1),
        statistics.median(list1),
        statistics.mode(list1)
        ]

a,b,c=mean_median_mode([3,4,2,2,3,3,2756,43])
print(f'Mean is {a} Median is {b} Mode is {c}')
````

### Multiple return statements 
Multiple return statements are allowed, but only the first return reached during a particular function call executes.

````python
def add(a,b):
    if a==0 & b==0:
        return "both zeros"
    else:
        return a+b
var1=int(input("enter first element"))
var2=int(input("Enter second element"))  
result=add(var1,var2)
print(result)
````
