### Debugging

- Debugging is the process of finding and fixing errors (bugs) in a program.
- The goal of debugging is to make the program work as expected.
- Errors can occur due to syntax mistakes, logical mistakes, or runtime issues.
- A person who debugs code is called a debugger (developer/programmer), and tools used for debugging are called debuggers.

#### Types of Errors

1. Syntax Errors
   - Violations of Python grammar.
   - Example:
     ```python
     if x > 5
         print(x)
     ```

2. Runtime Errors
   - Occur while the program is running.
   - Example:
     ```python
     print(10 / 0)
     ```
     (ZeroDivisionError)

3. Logical Errors
   - Program runs without crashing but produces incorrect output.
   - Example:
     ```python
     num = 10
     if num > 5:
         print("Less than 5")
     ```

### Common Python Errors (Exceptions)

1. SyntaxError
   - Invalid Python syntax.

2. IndentationError
   - Incorrect indentation.

3. NameError
   - Variable/function name not defined.

4. TypeError
   - Operation between incompatible types.

5. ValueError
   - Correct type but invalid value.

6. IndexError
   - List/Tuple index out of range.

7. KeyError
   - Dictionary key not found.

8. AttributeError
   - Object does not have the specified attribute/method.

9. ZeroDivisionError
   - Division by zero.

10. FileNotFoundError
    - File does not exist.

11. ModuleNotFoundError
    - Module cannot be imported.

12. ImportError
    - Import operation failed.

13. UnboundLocalError
    - Local variable referenced before assignment.

14. OverflowError
    - Result too large to represent.

15. MemoryError
    - Program runs out of memory.

16. RecursionError
    - Maximum recursion depth exceeded.

17. StopIteration
    - Iterator has no more items.

18. EOFError
    - Input reaches end-of-file unexpectedly.

19. KeyboardInterrupt
    - Program interrupted by user (Ctrl + C).

20. AssertionError
    - Assertion condition failed.

21. RuntimeError
    - Generic runtime error.

22. NotImplementedError
    - Method/function not implemented.

23. PermissionError
    - No permission to access resource.

24. OSError
    - Operating system related error.

25. TimeoutError
    - Operation timed out.
