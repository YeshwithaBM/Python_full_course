### Types of Arguments in Python

Arguments are values passed to a function when it is called. Python supports five types of arguments:

1. **Positional Arguments** – Values are matched to parameters based on their order. Example: `func(10, 20)`.
2. **Keyword Arguments** – Values are assigned using parameter names, so order doesn't matter. Example: `func(x=10, y=20)`.
3. **Default Arguments** – Parameters have default values that are used if no argument is provided. Example: `def func(x=10)`.
4. **Variable-Length Positional Arguments (`*args`)** – Accepts any number of positional arguments and stores them as a **tuple**.
5. **Variable-Length Keyword Arguments (`**kwargs`)** – Accepts any number of keyword arguments and stores them as a **dictionary**.

### Rules
- Parameter order should be: **Positional → Default → `*args` → `**kwargs`**.
- While calling a function, **positional arguments must come before keyword arguments**.
- `*args` and `**kwargs` are used to create flexible functions that can accept varying numbers of arguments.

### 1. Positional Arguments

```python
def student(name, age):
    print(f"Name: {name}")
    print(f"Age : {age}")

student("Yeshwitha", 22) #positional arguments
student(22,"yeshwitha") #makes no sense , Because Python only looks at the position. it doesn't follow position of parameters provided
```

### 2. Keyword Arguments

```python
def student(name, age):
    print(f"Name: {name}")
    print(f"Age : {age}")

student(age=22, name="Yeshwitha") #Notice the order doesn't matter.
```

### 3. Default Arguments

```python
def employee(name, country="India"): #Suppose most users are from India so we provide a default value .
    print(name,country)
    
employee("Rahul") #if no value passed or country , by default it takes India as its value 
employee("John", "USA") # if provided , it Override the default value
```

### Arbitrary arguments 
- Arbitrary Arguments is another name for Variable-Length Arguments.
##### There are two types of arbitrary arguments.
1. Arbitrary Positional Arguments (*args)


```python
# why tuples because its immutable ,values cannot be changed 
def add(*numbers): #stores in tuples (2,3,4,5,4,4,43)
    c=0
    print(numbers[0])
    #numbers[2]=3 #error
    for i in range(len(numbers)):
        c+=i
    print(f"sum of numbers is {c}")
add(2,3,4,5,4,4,43) #can provide any number of arguments , arguments in the sense positional arguments.
```

```python
def add(*numbers, name): #Python treats all positional arguments as part of numbers.
    c = 0
    print(numbers)
    print(name)
    for i in numbers:
        c += i

    print(f"Sum is {c}")

#add(1, 2, "Jenny") # one arguement missing
#to call this function correctly, name must be passed as a keyword argument:
add(1,2,4,name="yeshu")

#def func(name, *args): → name is positional.
#def func(*args, name): → name becomes keyword-only, so you must call it as name="Jenny".
```

2. Arbitrary keyword argument
- **kwargs 
- **kwargs allows a function to accept any number of keyword arguments. Python collects them into a dictionary.

```python
def student(**details): #stores as a dictionaries 
    print(details)

student(name="Yeshwitha", age=22, city="Bangalore") 
```

```python
# accessing values 
def student(**details):
    print(details["name"])
    print(details["age"])

student(name="Yeshwitha", age=22)
```

```python
#looping through kwargs
def student(**details):
    for key, value in details.items(): #items() is a dictionary method.
        print(key, ":", value)#.items() returns both the key and value of a dictionary.

student(
    name="Yeshwitha",
    age=22,
    city="Bangalore",
    role="AI Engineer"
)
```

Python always matches arguments in this sequence:

1. Normal positional parameters
        ↓
2. *args
        ↓
3. Keyword-only parameters
        ↓
4. **kwargs

```python
# ❌ Incorrect Function Definition
# Python always expects *args before **kwargs.
# Reason:
# Python matches arguments in the above order:
# Placing **kwargs before *args violates Python's function parameter order
# and results in a SyntaxError.

def info_person(**kwargs, *args):
    print("Args:", args)

    print("Kwargs:")
    for key, value in kwargs.items():
        print(key, value)

# ❌ Incorrect Function Call
# Positional arguments (1, 2) cannot appear after keyword arguments.
# In Python, all positional arguments must be passed before any keyword arguments.

info_person(
    name="Ram",
    age=30,
    dept="CSE",
    1,
    2
)
```

```python
#correct way to implement
def info_person(*args, **kwargs):
    print("Args:", args)

    print("Kwargs:")
    for key, value in kwargs.items():
        print(key, value)

info_person(
    1,
    2,
    name="Ram",
    age=30,
    dept="CSE"
)
```
