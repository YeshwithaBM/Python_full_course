### Local vs Global Scope

Variable = the actual named storage (x, age, name)
Scope = the region where that variable is visible and accessible.

- A variable declared inside a function is called a **local variable** and is created in the **local scope**.
- A variable declared outside all functions is called a **global variable** and is created in the **global scope**.
- Local variables can be accessed only inside the function where they are created.
- Global variables can be accessed both outside functions and inside functions (for reading).
- A local variable exists only while the function is executing.
- If a local variable and a global variable have the same name, the local variable takes precedence inside the function.
- To modify a global variable inside a function, the `global` keyword is required.

```python
a=10 #global scope 
def display():
    a=15 #local scope
    print(a)
display()
```

### Global Keyword

- `global` is a Python keyword.
- It is used inside a function to refer to a global variable.
- Global variables can be accessed (read) inside functions without using the `global` keyword.
- However, to modify a global variable inside a function, the `global` keyword is required.
- The `global` keyword tells Python that we are referring to the global variable, not creating a new local variable.
- Without `global`, assigning a value inside a function creates a local variable with the same name.

```python
#without global
count = 10

def update():
    count += 1
update()
```

```python
#with global keyword
count = 10

def update():
    global count
    count += 1

update()
print(count)
```

```python
a = 10   # global

def display():
    a = 20

    def show():
        global a
        a = 30
        print(a)
    show()
display()
print(a)
```
