## type hints 
Type hints allow us to specify what types of data our variables, function parameters and return values are expected to have.

- **Type Hints:** Specify expected data types for variables, function parameters, and return values.
- **Basic Types:** `int`, `float`, `str`, `bool`.
- **Collection Types:** `list`, `tuple`, `dict`, `set` with specified element types.
- **Union Types:** Allow a value to have more than one possible type.
- **Optional Types:** Indicate that a value can also be `None`.
- **Type Aliases:** Give a reusable name to a type definition.
- **Generics:** Allow functions and classes to work with different data types while preserving type information.
- **Dataclass:** A decorator that automatically generates methods like `__init__()` for classes mainly used to store data.
- **MyPy:** A static type checker that identifies type inconsistencies without executing the program.

**Remember:** Type hints improve readability and maintainability but are not enforced by Python at runtime.

```python
# variable type hints 
#colons are  called type annotations.
name: str = "yeshu"
age: int =23
name: str = "Alex"
age: int = 25
salary: float = 35000.50
is_active: bool = True
```

```python
#Type hints in collections
names: list[str] = ["Alex", "John"]

marks: list[int] = [80, 90, 95]

student: dict[str, int] = {
    "Alex": 90,
    "John": 85
}

coordinates: tuple[int, int] = (10, 20)

unique_ids: set[int] = {101, 102, 103}
```

```python
#Type hints in functions
#age: int means we expect age to be an integer.
#-> int means we expect the function to return an integer.
def create_user(first_name : str , last_name: str , age: int | None =None) -> dict[str,str | int | None] :
    email=f'{first_name.lower()}_{last_name.lower()}@gmail.com'

    return { 
        "first_name": first_name,
        "last_name" : last_name,
        "email": email,
        "age": age,
            }

user_1=create_user(name,"gowda",age)
print(user_1)
```

```python
from typing import NewType
from dataclasses import dataclass
@dataclass
class User:
    first_name: str
    last_name:str
    email: str 
    age: int | None = None
```

```python
def new_user(first_name : str , last_name: str , age: int | None =None) -> User :
    email=f'{first_name.lower()}_{last_name.lower()}@gmail.com'

    return User(
        first_name= first_name,
        last_name = last_name,
        email=  email,
        age= age,
    )

user_1=new_user("yeshu","gowda",'12')
print(user_1)
```

```python
#Type aliases
'''Suppose you repeatedly use a complicated type.
Instead of writing it repeatedly, you can give it another name.'''

StudentMarks = dict[str, int]
marks: StudentMarks = {
    "Alex": 90,
    "John": 85
}
#StudentMarks is a type alias. It gives a reusable name to an existing type.
```

```python
#Generics
'''Generics allow us to write reusable functions or classes 
that work with different data types while preserving type information.'''
#This works for integers, but what if we also want it to work for strings?
def get_first(items: list[int]) -> int:
    return items[0]
#We can use a generic type variable.
from typing import TypeVar
#The type variable T represents a type that is determined by the input.
T = TypeVar("T")

def get_first(items: list[T]) -> T:
    return items[0]
get_first([10, 20, 30])
get_first(["Alex", "John"])
```

```python
#Dataclasses
'''Sometimes we create classes mainly to store structured information.'''
#Normally, we write an initializer ourselves:
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age
```

```python
#Using a dataclass, Python generates 
#common methods such as __init__() automatically.
from dataclasses import dataclass

@dataclass
class Student:
    name: str
    age: int
    
student = Student("Alex", 22)

print(student.name)
print(student.age)
#Importantly, these annotations don't enforce the types at runtime.
```
