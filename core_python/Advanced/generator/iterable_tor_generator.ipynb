 
## Python: Iterables and Iterators
## 1. Iterable
- An iterable is an object that can provide an iterator to access its elements one at a time.
- Examples: List, Tuple, Set, Dictionary, String, Range.
- Integers, floats and booleans are not iterable by default.

Example:
    numbers = [10, 20, 30]

    iterator = iter(numbers)

## 2. Iterator
- An iterator is an object that produces elements one at a time.
- It remembers its current position (state).
- It moves forward whenever we request the next element.
- Once exhausted, it raises StopIteration.

Example:
    numbers = [10, 20, 30]

    iterator = iter(numbers)

    print(next(iterator))  # 10
    print(next(iterator))  # 20
    print(next(iterator))  # 30

## 3. iter() Function
- `iter()` obtains an iterator from an iterable.
- Internally, it calls the object's `__iter__()` method.

Syntax:
    iterator = iter(iterable)

Example:
    numbers = [10, 20, 30]
    iterator = iter(numbers)

**Remember:** `iter()` creates/obtains the iterator; it does not retrieve the first element.

## 4. next() Function
- `next()` requests the next element from an iterator.
- Internally, it calls the iterator's `__next__()` method.
- Each call advances the iterator's internal state.

Syntax:
    next(iterator)
    next(iterator, default_value)

Example:
    numbers = [10, 20]

    iterator = iter(numbers)

    print(next(iterator))  # 10
    print(next(iterator))  # 20
    print(next(iterator, "Finished"))  # Finished

## 5. __iter__() and __next__()
These are Python's special (dunder) methods.

- `__iter__()` returns an iterator.
- `__next__()` produces the next element.
- When no more elements exist, `__next__()` raises `StopIteration`.

## 6. StopIteration
- An exception raised when an iterator has no more elements.
- It signals that iteration is complete.
- A `for` loop handles this exception automatically.
- Use the second argument of `next()` to provide a default value instead.

## 7. Iterable vs Iterator

| Iterable | Iterator |
|---|---|
| Provides an iterator | Produces elements |
| Contains or provides access to data | Remembers its current state |
| Can create multiple iterators | Tracks its own progress |
| Example: List | Example: List iterator |

**Important:** Every iterator is also an iterable, but not every iterable is an iterator.

## 8. Can an Iterator Be an Iterable?

Yes!

An iterator implements both `__iter__()` and `__next__()`.

Example:
    numbers = [10, 20, 30]

    iterator = iter(numbers)
    lol = iter(iterator)

    print(iterator is lol)  # True

**Why True?**
- Calling `iter()` on an existing iterator returns the same iterator.
- It does not create another iterator.
- Both variables reference the same object and share the same position.

## 9. Multiple Iterators from the Same List

    numbers = [10, 20, 30]

    a = iter(numbers)
    b = iter(numbers)

    print(next(a))  # 10
    print(next(a))  # 20
    print(next(b))  # 10

**Why?**
- `a` and `b` are separate iterator objects.
- Each maintains its own position.
- Moving one iterator does not move the other.

## 10. How Does a for Loop Work?

A `for` loop automatically uses the iterator protocol.

    numbers = [10, 20, 30]

    for n in numbers:
        print(n)

Conceptually, Python:
1. Calls `iter(numbers)`.
2. Calls `next()` repeatedly.
3. Executes the loop body for each value.
4. Stops when `StopIteration` occurs.

## 11. Why Do We Use Iterators?
- Memory efficiency: Values can be processed one at a time.
- Lazy evaluation: Values are generated when requested.
- Large data processing: Useful for files, datasets and streams.
- Data processing: Used by `map()`, `filter()` and generators.

## 12. Important Terms to Remember

| Term | Meaning |
|---|---|
| Iterable | An object that can provide an iterator |
| Iterator | An object that produces values one at a time |
| `iter()` | Obtains an iterator |
| `next()` | Requests the next value |
| `__iter__()` | Special method for obtaining an iterator |

#### Python Generators – Brief Notes

## 1. What is a Generator?
- A generator is a special type of **iterator**.
- It produces values one at a time instead of storing all values in memory.
- Generators are commonly created using a function with the `yield` keyword.

## 2. What is `yield`?
- `yield` returns a value and temporarily pauses the function.
- It remembers its execution state.
- When the next value is requested, execution resumes from where it paused.

Example:

    def numbers():
        yield 10
        yield 20
        yield 30

    gen = numbers()

Calling `numbers()` creates a generator object. The function body does not execute immediately.

## 3. How does `next()` work?
- `next()` requests the next value from a generator.
- Each call resumes execution until the next `yield`.
- Once all values are exhausted, it raises `StopIteration`.

    print(next(gen))  # 10
    print(next(gen))  # 20
    print(next(gen))  # 30

## 4. Generator with a `for` loop

A `for` loop automatically:
1. Gets the iterator using `iter()`.
2. Calls `next()` repeatedly.
3. Stops when `StopIteration` occurs.

    def square_numbers(nums):
        for i in nums:
            yield i * i

    gen = square_numbers([1, 2, 3, 4])

    for value in gen:
        print(value)

Output:

    1
    4
    9
    16

**Note:** A generator maintains its position. If you partially consume it using `next()`, a `for` loop continues from where it stopped.

## 5. `return` vs `yield`

| `return` | `yield` |
|---|---|
| Returns a value and exits the function. | Produces a value and pauses the function. |
| Function execution ends. | Function execution can resume. |
| Does not preserve execution for another call. | Preserves execution state. |

## 6. List vs Generator

| List | Generator |
|---|---|
| Creates and stores all elements. | Produces elements on demand. |
| Consumes more memory for large collections. | Usually consumes less memory. |
| Supports indexing. | Does not support direct indexing. |
| Can be iterated multiple times. | Normally exhausted after one iteration. |
| Useful when you need all values available. | Useful for lazy processing of large data. |

## 7. Why are Generators Useful?
- Memory-efficient processing.
- Lazy evaluation (values are generated only when requested).
- Useful for large datasets.
- Useful for reading large files line by line.
- Useful for processing database records and streaming data.

**Important:** Generators are not necessarily faster than lists. Their main advantage is memory efficiency and on-demand processing.

```python

import mem_profile
import random
import time

names = ['John', 'Corey', 'Adam', 'Steve', 'Rick', 'Thomas']
majors = ['Math', 'Engineering', 'CompSci', 'Arts', 'Business']

print('Memory (Before): {}Mb'.format(
    mem_profile.memory_usage_psutil()
))


def people_list(num_people):
    result = []

    for i in range(num_people):
        person = {
            'id': i,
            'name': random.choice(names),
            'major': random.choice(majors)
        }

        result.append(person)

    return result


def people_generator(num_people):
    for i in range(num_people):
        person = {
            'id': i,
            'name': random.choice(names),
            'major': random.choice(majors)
        }

        yield person


t1 = time.perf_counter()
people = people_list(1000000)
t2 = time.perf_counter()

# t1 = time.perf_counter()
# people = people_generator(1000000)
# t2 = time.perf_counter()

print('Memory (After): {}Mb'.format(
    mem_profile.memory_usage_psutil()
))

print('Took {} Seconds'.format(t2 - t1))
```
