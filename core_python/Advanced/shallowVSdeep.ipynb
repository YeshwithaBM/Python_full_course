### shallow type and deep type 
- A shallow copy creates a new outer container but copies references to the objects inside it. A deep copy recursively creates copies of the nested objects too.
#### Differences Between Assignment, Shallow Copy and Deep Copy in Python

| Feature | Normal Assignment | Shallow Copy | Deep Copy |
|---|---|---|---|
| Syntax | `b = a` | `copy.copy(a)` | `copy.deepcopy(a)` |
| New outer list? | No | Yes | Yes |
| Outer objects | Same object | Separate objects | Separate objects |
| Nested objects | Shared | Shared | Recursively copied |
| Outer list modification | Affects both | Affects only the modified list | Affects only the modified list |
| Nested list modification | Affects both | Affects both | Does not affect the other |
| References | Both refer to the same object | New outer list, shared inner references | Separate copied object structure |

##### Simple Explanation

1. **Normal Assignment:** Creates another reference to the same object. Both variables point to the same list.

2. **Shallow Copy:** Creates a new outer list, but the nested lists are shared through references. Changes to a nested list affect both copies.

3. **Deep Copy:** Creates a new outer list and recursively copies nested objects. Changes in one copy do not affect the other.

**Remember:**
- Assignment = Same object.
- Shallow Copy = New outer object, shared nested objects.
- Deep Copy = New outer and recursively copied nested objects.
