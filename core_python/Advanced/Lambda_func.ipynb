# Python: Lambda, Map and Filter

## 1. Lambda Function
- A `lambda` function is a small, anonymous function (a function without a name).
- Used for short, simple operations.
- Can contain only one expression.
- Automatically returns the result; no `return` keyword is required.

**Syntax:**
    lambda arguments: expression

**Example:**
    square = lambda n: n * n
    print(square(5))  # 25

Equivalent using `def`:
    def square(n):
        return n * n

## 2. map() Function
- Used to **transform every element** in an iterable.
- Applies a given function to each element.
- Returns a map iterator, not a list.
- Use `list()` to collect the results into a list.

**Syntax:**
    map(function, iterable)

**Example:**
    numbers = [1, 2, 3, 4]

    def square(n):
        return n * n

    result = list(map(square, numbers))
    print(result)  # [1, 4, 9, 16]

## 3. filter() Function
- Used to **select elements** based on a condition.
- Keeps elements for which the function returns `True`.
- Discards elements for which it returns `False`.
- Returns a filter iterator, not a list.
- Use `list()` to collect the results.

**Syntax:**
    filter(function, iterable)

**Example:**
    numbers = [1, 2, 3, 4, 5, 6]

    def is_even(n):
        return n % 2 == 0

    result = list(filter(is_even, numbers))
    print(result)  # [2, 4, 6]

## 4. Difference Between map() and filter()

| Feature | map() | filter() |
|---|---|---|
| Purpose | Transform elements | Select elements |
| Function returns | Transformed value | True or False |
| Output count | Same as input | Zero to input count |
| Returns | Map iterator | Filter iterator |

## Remember

- `lambda` → Creates a small anonymous function.
- `map()` → What should I do to every element?
- `filter()` → Which elements should I keep?
- `list()` → Collects iterator results into a list.

**Important:** Both `map()` and `filter()` accept a function and an iterable. They do not automatically modify the original iterable.

```python
## Lambda filter 
prices=[100,120,123,30,26,78]
print(list(filter(lambda p: p>=100, prices)))
```

```python
students=[["maria",85],["kumra",90],["max",67]]
print(list(filter(lambda row: row[0].startswith("m") , students)))
```

```python
'''Suppose you want to:
1. Select prices greater than ₹500.
2. Apply a 10% discount to those prices.'''
prices = [100, 250, 500, 1000, 2000]

def expensive(price):
    return price > 500

def discount(price):
    return price * 0.90

result = list(map(discount, filter(expensive, prices)))

print(result)
```
