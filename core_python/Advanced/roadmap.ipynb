## 3. Important Workflows to Understand

### A. AI Request Workflow

1. Browser sends a request.
2. `AIUser` checks login and limits.
3. `@metered` reserves credits.
4. Prompt builder wraps the user's text.
5. `llm_provider` calls the LLM with retry handling.
6. Parser processes the generated output.
7. Backend sends the response to the browser.

**Flow:**

`Browser → AIUser → @metered → Prompt Builder → LLM Provider → Parser → Response`

### B. Login Workflow

1. User enters their password.
2. Optional 2FA verification takes place.
3. JWT is generated.
4. Every subsequent request carries the token.
5. `token_version` is checked on each request.

**Flow:**

`Password → Optional 2FA → JWT → Authenticated Requests → token_version Check`

### C. Blueprint Workflow

1. User submits a description.
2. The description is analysed.
3. Follow-up questions are generated.
4. A plan is created.

### D. Export and Deployer Workflow

1. Canvas elements are collected.
2. `exportFiles.js` converts them into HTML and JSX.
3. `index.html` and JSX files are generated.
4. Deployer publishes `index.html` as a live website.

**Important:** Untrusted text must be escaped to prevent code injection and other security issues.

### E. Git Workflow

1. Create a branch from `origin/test`.
2. Make code changes.
3. Commit changes.
4. Push the branch.
5. Create a pull request into `test`.
6. Fix review comments.
7. Push additional commits to update the same PR.

**Flow:**

`origin/test → New Branch → Commit → Push → PR → Review → Fixes`
