 
## Python: Iterables and Iterators

## 1. Iterable
- An iterable is an object that can provide an iterator to access its elements one at a time.
- Examples: List, Tuple, Set, Dictionary, String, Range.
- Integers, floats and booleans are not iterable by default.

Example:
    numbers = [10, 20, 30]

    iterator = iter(numbers)

## 2. Iterator
- An iterator is an object that produces elements one at a time.
- It remembers its current position (state).
- It moves forward whenever we request the next element.
- Once exhausted, it raises StopIteration.

Example:
    numbers = [10, 20, 30]

    iterator = iter(numbers)

    print(next(iterator))  # 10
    print(next(iterator))  # 20
    print(next(iterator))  # 30

## 3. iter() Function
- `iter()` obtains an iterator from an iterable.
- Internally, it calls the object's `__iter__()` method.

Syntax:
    iterator = iter(iterable)

Example:
    numbers = [10, 20, 30]
    iterator = iter(numbers)

**Remember:** `iter()` creates/obtains the iterator; it does not retrieve the first element.

## 4. next() Function
- `next()` requests the next element from an iterator.
- Internally, it calls the iterator's `__next__()` method.
- Each call advances the iterator's internal state.

Syntax:
    next(iterator)
    next(iterator, default_value)

Example:
    numbers = [10, 20]

    iterator = iter(numbers)

    print(next(iterator))  # 10
    print(next(iterator))  # 20
    print(next(iterator, "Finished"))  # Finished

## 5. __iter__() and __next__()
These are Python's special (dunder) methods.

- `__iter__()` returns an iterator.
- `__next__()` produces the next element.
- When no more elements exist, `__next__()` raises `StopIteration`.

## 6. StopIteration
- An exception raised when an iterator has no more elements.
- It signals that iteration is complete.
- A `for` loop handles this exception automatically.
- Use the second argument of `next()` to provide a default value instead.

## 7. Iterable vs Iterator

| Iterable | Iterator |
|---|---|
| Provides an iterator | Produces elements |
| Contains or provides access to data | Remembers its current state |
| Can create multiple iterators | Tracks its own progress |
| Example: List | Example: List iterator |

**Important:** Every iterator is also an iterable, but not every iterable is an iterator.

## 8. Can an Iterator Be an Iterable?

Yes!

An iterator implements both `__iter__()` and `__next__()`.

Example:
    numbers = [10, 20, 30]

    iterator = iter(numbers)
    lol = iter(iterator)

    print(iterator is lol)  # True

**Why True?**
- Calling `iter()` on an existing iterator returns the same iterator.
- It does not create another iterator.
- Both variables reference the same object and share the same position.

## 9. Multiple Iterators from the Same List

    numbers = [10, 20, 30]

    a = iter(numbers)
    b = iter(numbers)

    print(next(a))  # 10
    print(next(a))  # 20
    print(next(b))  # 10

**Why?**
- `a` and `b` are separate iterator objects.
- Each maintains its own position.
- Moving one iterator does not move the other.

## 10. How Does a for Loop Work?

A `for` loop automatically uses the iterator protocol.

    numbers = [10, 20, 30]

    for n in numbers:
        print(n)

Conceptually, Python:
1. Calls `iter(numbers)`.
2. Calls `next()` repeatedly.
3. Executes the loop body for each value.
4. Stops when `StopIteration` occurs.

## 11. Why Do We Use Iterators?
- Memory efficiency: Values can be processed one at a time.
- Lazy evaluation: Values are generated when requested.
- Large data processing: Useful for files, datasets and streams.
- Data processing: Used by `map()`, `filter()` and generators.

## 12. Important Terms to Remember

| Term | Meaning |
|---|---|
| Iterable | An object that can provide an iterator |
| Iterator | An object that produces values one at a time |
| `iter()` | Obtains an iterator |
| `next()` | Requests the next value |
| `__iter__()` | Special method for obtaining an iterator |
| `__next__()` | Special method for producing the next value |
| `StopIteration` | Signals the end of iteration |
| Lazy evaluation | Producing values only when requested |

### The Complete Concept

Iterable → `iter()` → Iterator → `next()` → Next element

**In simple words:**

An iterable provides the data, an iterator remembers where it is, and `next()` retrieves the next element by advancing the iterator's state.

```python
numbers = [10, 20, 30]

iterator = iter(numbers)  # Creates an iterator

value = next(iterator)    # Retrieves the next value

print(value)
```

```python
from collections.abc import Iterable, Iterator

numbers = [10, 20, 30]
iterator = iter(numbers)

print(isinstance(numbers, Iterable))  # True
print(isinstance(numbers, Iterator))  # False

print(isinstance(iterator, Iterable))  # True
print(isinstance(iterator, Iterator))  # True
```
