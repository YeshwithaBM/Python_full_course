### Python – Exception Handling

## 1. What is Exception Handling?
Exception handling is a mechanism used to handle runtime exceptions so that a program doesn't terminate unexpectedly.

## 2. try Block
- Contains code that might raise an exception.
- Python executes the code inside it.
- If an exception occurs, the remaining statements in the try block are skipped.

## 3. except Block
- Handles exceptions raised inside the try block.
- We can specify particular exception types.
- `as e` allows us to access the exception details.

Example:

    try:
        result = 10 / 0
    except ZeroDivisionError as e:
        print("Error:", e)

## 4. Multiple except Blocks
We can handle different exceptions separately.

- `except ValueError`: Handles ValueError.
- `except TypeError`: Handles TypeError.
- `except Exception`: Handles most ordinary exceptions.

**Important:** Always place specific exception handlers before the general `except Exception` handler.

## 5. else Block
- Executes only when the try block completes without an exception.
- Used for operations that should happen after successful execution.

## 6. finally Block
- Executes during normal or exceptional exit.
- Commonly used for cleanup operations.
- Examples: Closing files, database connections and releasing resources.

## 7. raise Keyword
- Used to explicitly raise an exception.
- Useful when we want to enforce our own validation rules.

Example:

    age = -5

    if age < 0:
        raise ValueError("Age cannot be negative")

## 8. Types of Errors

| Type | Meaning | Example |
|---|---|---|
| Syntax Error | Violates Python syntax rules. | Missing colon |
| Runtime Error / Exception | Occurs while the program executes. | Division by zero |
| Logical Error | Program executes but produces an incorrect result. | Using subtraction instead of addition |

## 9. Can all errors be handled using try-except?

- Runtime exceptions can be handled using try-except.
- Syntax errors in the code itself are detected before normal execution and cannot be caught by a try block containing that invalid syntax.
- Syntax errors raised during dynamic execution, such as `exec()`, can be caught.
- Logical errors are generally identified through debugging and testing.

## 10. Common Python Exceptions

- `ValueError` – Invalid value.
- `TypeError` – Invalid operation between incompatible types.
- `ZeroDivisionError` – Division by zero.
- `IndexError` – Invalid list index.
- `KeyError` – Missing dictionary key.
- `FileNotFoundError` – File does not exist.
- `ImportError` – Import operation fails.
- `SyntaxError` – Invalid Python syntax.

## Quick Revision

| Keyword | Purpose |
|---|---|
| `try` | Contains code that might raise an exception. |
| `except` | Handles an exception. |
| `else` | Executes when no exception occurs in try. |
| `finally` | Executes during normal or exceptional exit; used for cleanup. |
| `raise` | Explicitly triggers an exception. |
| `as e` | Stores the exception object in a variable. |

**Remember:** Exception handling helps applications handle unexpected situations gracefully instead of terminating abruptly.

```python
### Try else finally 
'''finally block always gets executed even if either else or 
exception handling block executes 
when we open the connection for request then finally acts as closing the connection
'''
try:
    num1=int(input("enter a number:"))
    num2=int(input("enter another Number:"))
    c=num1/num2
    d=num1*num2
    e=num1+num2
except ZeroDivisionError as z :
    print(z)
except ValueError as v:
    print(v)
except Exception as e: # so it executes 
    print(e)
else:
    print(c)
    print(d)
    print(e)
finally:
    print("Execution Successful")
```

### Custom Exception Handling 

```python
def calculate_grade(marks):
    if marks < 0 or marks > 100:
        raise ValueError("Invalid marks")

    return marks

print(calculate_grade(150))
```

```python
def place_order(quantity):
    if quantity <= 0:
        raise ValueError("Quantity must be greater than zero")

    print("Order placed")

place_order(-2)
```

```python
#Create your own exception
#Suppose you want to raise an error called InvalidPasswordError.
class InvalidPasswordError(Exception):
    pass
```

```python
def create_password(password):
    if len(password) < 8:
        raise InvalidPasswordError("Password is too short")

    print("Password created successfully")

create_password("abc")
```

```python
try:
    create_password("abc")

except InvalidPasswordError as error:
    print(error)
```
