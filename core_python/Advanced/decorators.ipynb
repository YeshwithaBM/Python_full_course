### Python – Function Copy, Closures & Decorators

### 1. Function Copy (First-Class Functions)

In Python, functions are **first-class objects**. This means functions can be:

- Assigned to variables.
- Passed as arguments to other functions.
- Returned from other functions.
- Stored in data structures.

### Example 1: Assigning a function to another variable

    def greet():
        print("Hello!")

    say_hello = greet

    say_hello()

- `say_hello` and `greet` refer to the same function object.
- No new function is created.
- `greet` can be deleted using `del greet`, but `say_hello()` will still work because another reference exists.

### Example 2: Returning a function

    def greet():
        print("Hello!")

    def get_function():
        return greet

    my_function = get_function()
    my_function()

- `return greet` returns the function object itself.
- `return greet()` executes the function and returns its result.
- A function object can still be used after its original name is deleted, as long as another reference exists.

**Important:** Deleting a variable name does not necessarily delete the object it refers to.


### 2. Closures

A **closure** is a function that remembers and accesses variables from its enclosing function, even after the enclosing function has finished executing.

### Example:

    def outer():
        message = "Hello"

        def inner():
            print(message)

        return inner

    my_function = outer()
    my_function()

Output:

    Hello

### How does it work?

1. `outer()` executes and creates the variable `message`.
2. The inner function uses the outer function's variable.
3. The inner function is returned.
4. The outer function finishes execution.
5. The returned inner function can still access `message`.

### Conditions for a Closure

1. A nested function must exist.
2. The inner function must use a variable from its enclosing scope.
3. The inner function must retain access to that variable after the outer function finishes.

### Important Points

- Closures preserve access to enclosing variables.
- They can maintain state between function calls.
- Multiple closures can remember different values.
- `nonlocal` allows an inner function to modify a variable in its enclosing scope.
- Returning the inner function is common, but not mandatory for a closure.

### Example: Counter using Closure

    def counter():
        count = 0

        def increment():
            nonlocal count
            count += 1
            print(count)

        return increment

    my_counter = counter()

    my_counter()  # 1
    my_counter()  # 2
    my_counter()  # 3

### 3. Decorators

A **decorator** is a function that takes another function as an argument and returns a function that adds or modifies its behavior without changing its original implementation.

### Why do we use decorators?

- To add extra functionality to existing functions.
- To avoid repeating the same code.
- To execute additional operations before or after a function.
- Commonly used for logging, authentication, timing and validation.

### Basic Decorator Example

    def decorator(original_function):

        def wrapper():
            print("Before function execution")

            original_function()

            print("After function execution")

        return wrapper

    def greet():
        print("Hello!")

    greet = decorator(greet)

    greet()

Output:

    Before function execution
    Hello!
    After function execution

### How does it work?

1. The original function is passed to the decorator.
2. The decorator defines a wrapper function.
3. The wrapper contains additional functionality.
4. The wrapper calls the original function.
5. The decorator returns the wrapper function.
6. The original function name now refers to the wrapper.

### Decorator Syntax: `@`

Instead of writing:

    greet = decorator(greet)

We can write:

    @decorator
    def greet():
        print("Hello!")

Both approaches are equivalent.

### Decorators with Arguments

Use `*args` and `**kwargs` so the wrapper can accept different types and numbers of arguments.

    def decorator(original_function):

        def wrapper(*args, **kwargs):
            print("Before execution")

            result = original_function(*args, **kwargs)

            print("After execution")

            return result

        return wrapper

### Important Points

- A decorator typically uses first-class functions and closures.
- The wrapper adds extra functionality around the original function.
- `*args` accepts positional arguments.
- `**kwargs` accepts keyword arguments.
- `return result` preserves the original function's return value.
- The decorator is applied when Python processes the decorated function definition.
- The wrapper executes when the decorated function is called.

### Real-world Decorator Examples

| Decorator | Purpose |
|---|---|
| `@app.get()` | Registers a GET route in FastAPI. |
| `@app.post()` | Registers a POST route in FastAPI. |
| `@staticmethod` | Defines a static method. |
| `@classmethod` | Defines a class method. |
| `@property` | Allows a method to be accessed like an attribute. |

## 4. Quick Revision

| Concept | Remember |
|---|---|
| Function Copy / First-Class Functions | Functions can be assigned, passed and returned like other objects. |
| Closures | Inner functions retain access to enclosing variables. |
| Decorators | Add functionality to existing functions using a wrapper. |

### How are these three connected?

1. **First-class functions:** Allow us to pass and return functions.
2. **Closures:** Allow inner functions to remember enclosing variables.
3. **Decorators:** Use these concepts to wrap existing functions and add functionality.

**Most important takeaway:**

A decorator typically receives an original function, creates a wrapper that remembers it, and returns that wrapper to replace the original function reference.

```python
## decorators
def decorator(original_function):

    def wrapper():
        print("Before function execution")

        original_function()

        print("After function execution")

    return wrapper
```

```python
def greet():
    print("Hello!")

greet = decorator(greet)
greet()
```

```python
@decorator
def greet():
    print("Hello!")
greet()
```

```python
def decorator(original_function):

    def wrapper(*args, **kwargs):
        print("Before function execution")

        result = original_function(*args, **kwargs)

        print("After function execution")

        return result

    return wrapper


@decorator
def add(a, b):
    return a + b


answer = add(10, 20)

print(answer)
```
