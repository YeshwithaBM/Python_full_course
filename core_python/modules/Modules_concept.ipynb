### Modules

- **Modules** are Python files (`.py`) that contain reusable **code, functions, classes, or data**.
- module is a Python file containing reusable code or data that can be imported and used in another Python program.

- Modules are mainly used for:
  - **Code reusability** — write code once and use it in multiple programs.
  - **Code organization** — divide a large program into smaller, manageable files.
  - **Maintainability** — easier to understand, modify, and debug code.

- When we need functionality from a module, we can **import** it into our program and use its contents.

### Types of Modules

1. **Built-in / Standard Library Modules**
   - Modules that come with Python.
   - Examples:
     ```python
     import math
     import os
     import random
     ```

2. **User-Created Modules**
   - Modules created by the programmer.
   - Example:

     `calculator.py`
     ```python
     def add(a, b):
         return a + b
     ```

     `main.py`
     ```python
     import calculator

     print(calculator.add(10, 20))
     ```

### `import ... as ...`

- If the module name is long, instead of repeatedly writing the full module name, we can give it a **short alias** using `as`.

- `as` creates an **alias** for the imported module, making it shorter and easier to use.

- Here:

  - `my_module` → actual module name
  - `m` → alias (short name) given to the module

- Now instead of:

```python
my_module.some_function()
my_module.some_variable
```
- we can write :
```python
m.some_function()
m.some_variable

````python
import my_module as m
print(m.area_of_square(2))
````

````python
import random as r
print(r.randint(1,90))
````

### `from ... import ...`

- If we only need a **specific function, variable, or class** from a module, we can import only that specific thing.

```python
from my_module import my_function
```

- Now we can directly use the function without writing the module name:

```python
my_function()
```

---

### Importing Multiple Specific Things

We can import multiple things from the same module:

```python
from my_module import my_function, my_variable
```

Then:

```python
my_function()
print(my_variable)
```

---

### Giving an Alias to an Imported Thing

We can also give the imported function/variable an alias using `as`.

```python
from my_module import my_function as func
```

Now:

```python
func()
```

Here:

```text
my_function → original name
func        → alias
```

---

### `from ... import *`

- `*` means **import all names** from the module.

```python
from my_module import *
```

Now we can directly use the functions and variables:

```python
my_function()
print(my_variable)
```

- We don't need to write:

```python
my_module.my_function()
```

- However, using `import *` is generally avoided in larger programs because it can make it unclear where a name came from and can cause name conflicts.

---

### Name Conflict

Suppose our module contains:

```python
# my_module.py

a = 10

def show():
    print("Hello")
```

And our current file contains:

```python
a = 5
```

If we do:

```python
from my_module import *
```

then the imported `a` can overwrite the existing `a` in the current namespace.

```python
print(a)
```

This can cause confusion because there are now two possible sources for the name `a`.

---

### Safest Way to Access the Module's Specific Variable

Instead of importing everything directly, import the module itself:

```python
import my_module
```

Then use:

```python
print(my_module.a)
my_module.show()
```

Now the module's `a` is clearly identified.

We can also use an alias:

```python
import my_module as m

print(m.a)
m.show()
```

Here:

```text
my_module → actual module name
m         → alias
a         → variable inside my_module
```

### Important Difference

```python
from my_module import a
```

means:

```text
Bring `a` directly into the current namespace.
```

So we use:

```python
print(a)
```

Whereas:

```python
import my_module
```

means:

```text
Import the module itself.
```

So we use:

```python
print(my_module.a)
```

### Quick Summary

```text
import module
    ↓
module.function()
module.variable()

import module as m
    ↓
m.function()
m.variable()

from module import function
    ↓
function()

from module import function as f
    ↓
f()

from module import *
    ↓
function()
variable()
```

> **Main idea:** `import module` keeps the module namespace, while `from module import thing` brings the selected thing directly into the current namespace.
