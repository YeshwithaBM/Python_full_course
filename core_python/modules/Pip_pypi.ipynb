#### Pip & PyPI

##### Package

A **package** is a way of organizing related Python modules and subpackages together.

A package is generally represented by a **directory/folder** containing Python code.

A package is a way of organizing Python modules; a library is a broader term for reusable code, which may contain one or more packages/modules.
Example:

    mypackage/
        __init__.py
        calculator.py
        geometry.py
        utils.py

Here:

- `calculator.py` → module
- `geometry.py` → module
- `utils.py` → module
- `mypackage` → package

A package can also contain **subpackages**.

---

##### PyPI

**PyPI = Python Package Index**

PyPI is a public repository where Python developers can **publish and distribute Python packages**.

Developers publish reusable packages to PyPI, and other developers can find and install them.

---

##### pip

**pip** is Python's package installer.

It is used to:

- Install packages
- Upgrade packages
- Uninstall packages
- Manage Python package dependencies

Example:

    pip install requests

By default, pip looks for packages on **PyPI**.

So the basic relationship is:

    Developer
        ↓
    Creates package
        ↓
    Publishes package
        ↓
       PyPI
        ↓
    User uses pip
        ↓
    pip installs package

---

##### Simple Example

Suppose a developer creates a package called:

    mycalculator

They publish it to PyPI.

Another developer can install it using:

    pip install mycalculator

Then they can use the package in their Python project.

---

##### Easy Difference

**Package** → organized collection of related Python modules/code.

**PyPI** → online repository containing published Python packages.

**pip** → tool used to install/manage those packages.

##### Remember

> **PyPI stores/distributes packages, while pip installs and manages them.**
