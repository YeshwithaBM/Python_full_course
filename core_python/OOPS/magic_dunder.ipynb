# Magic Methods / Dunder Methods / Special Methods

**Magic methods, dunder methods, and special methods** are different names for Python's special methods.

They have **double underscores** before and after their name:

    __init__()
    __str__()
    __add__()
    __len__()
    __eq__()

## What are they?

> **Dunder methods are special methods that Python automatically calls to define how objects behave with built-in operations and features.**

## Why do we use them?

They allow our **custom objects to work naturally with Python syntax and built-in functions**.

For example:

    class Number:
        def __init__(self, value):
            self.value = value

        def __add__(self, other):
            return self.value + other.value

    n1 = Number(10)
    n2 = Number(20)

    print(n1 + n2)

When Python sees:

    n1 + n2

it uses:

    n1.__add__(n2)

So `__add__()` defines how `+` behaves for our objects.

## Common Dunder Methods

| Dunder Method | Used for |
|---|---|
| `__init__()` | Initialize an object |
| `__str__()` | String representation |
| `__add__()` | `+` |
| `__sub__()` | `-` |
| `__eq__()` | `==` |
| `__gt__()` | `>` |
| `__len__()` | `len()` |
| `__getitem__()` | `obj[index]` |

## Main Purpose

Dunder methods act as a **connection between our custom objects and Python's built-in operations**.

They allow us to define:

    obj1 + obj2
    obj1 == obj2
    len(obj)
    print(obj)
    obj[index]

for our own classes.

### Connection with OOP

Dunder methods allow objects of our classes to **behave like normal Python objects** and are heavily used for features such as **operator overloading and object initialization**.

```python
class Author:
    def __init__(self,name,book_name,pages):
        self.name=name
        self.book_name=book_name
        self.pages=pages
    def __len__(self):
        return self.pages
    def __str__(self):
        return f"Author name is {self.name} and book name is {self.book_name} and pages are {self.pages}"
    def __call__(self):
        print("HI, I am a callable object")
d=Author("John Doe", "Python Programming", 300)
# print(str(d))
# print(d.__str__())
print(d)
# print(d.__len__())
print(len(d))
d()
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def __gt__(self, other):
        if self.age > other.age :
            return True
        else:
            return False

p1=Person("mohan", 30)
p2=Person("yeshu", 25)
print(p1.__gt__(p2)) # here, the __gt__ method is called directly to compare the ages of two Person objects, resulting in the True boolean value.
print(p1 > p2) # here, the > operator is overloaded to compare the ages of two Person objects, resulting in the same True boolean value.
if p1 > p2:
    print(f"{p1.name} is paying the bill")
else:
    print(f"{p2.name} is paying the bill")
```
