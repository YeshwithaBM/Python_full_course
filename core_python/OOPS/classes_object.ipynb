### Classes and Objects

- A **class** is a design/blueprint used to create objects.
- OOP is used to **model real-world problems and entities as objects**.
- Every object can have some **information/data** and can perform some **actions/operations**.
- What an object **has** can be modeled using **attributes** (variables).
- What an object **does** can be modeled using **methods** (functions).
- Attributes represent the **data/state** of an object.
- Methods represent the **behavior/operations** of an object.

#### Class

A **class** is a **user-defined type** that specifies the **structure and behavior** that objects of that type can have.

A class can define:

- **Attributes** → the data/state associated with objects.
- **Methods** → the behavior/operations that objects can perform.

#### Object

- An **object** is a **particular entity of a class** that has its own **identity, state, and behavior**.
- Objects are created based on the structure and behavior defined by their class.
 
##### **The important thing is that the object is the actual car represented by that variable, not the variable name itself.**

- ** A class is a type. In Python, that type is itself represented by a class object.**

##### Entity
→ general word for a particular thing

##### Instance
→ programming/OOP term for an object belonging to a class/type

##### Example
```text
Class: Car

├── Attributes → color, speed

       ↓
