##### 🧬 Multilevel Inheritance & MRO in Python

---

#### 📌 1. What is Multilevel Inheritance?
**Multilevel inheritance** forms a **chain of inheritance** where a child class inherits from a parent class, which in turn inherits from its own parent class.

* **Chain:** `Grandparent (A)` ➔ `Parent (B)` ➔ `Child (C)`
* **Concept:** Class `C` directly inherits from `B` and **indirectly inherits** from `A`. Therefore, `C` accesses features of **both** `A` and `B`.

```python
class Human:
    pass

class Male(Human):    # Directly inherits from Human
    pass

class Boy(Male):      # Directly inherits from Male, indirectly from Human
    pass
```

---

##### 🔍 2. Method Resolution Order (MRO)
**MRO** defines the precise order Python follows to search for methods and attributes.

* **Search Priority:** Current class ➔ Parent class ➔ Grandparent class ➔ `object`
* **Early Stop:** Python uses the first matching method it finds and stops searching.

```python
print(Boy.mro())
# Output: [<class '__main__.Boy'>, <class '__main__.Male'>, <class '__main__.Human'>, <class 'object'>]
```

> **Note on `object`:** In Python 3, `object` is the ultimate built-in base class for all classes (even if not explicitly written).

---

#### ⚡ 3. Key Concepts: `super()` & Combining Types

### Using `super()`
`super()` delegates method calls to the **next class in the MRO**.
```python
class Boy(Male):
    def work(self):
        super().work()  # Calls work() from Male (or Human if not in Male)
```

### Multilevel vs. Multiple Inheritance
| Type | Structure | Description |
| :--- | :--- | :--- |
| **Multilevel** | `A ➔ B ➔ C` | Inherits through a chain of levels. |
| **Multiple** | `(A, B) ➔ C` | A single child directly inherits from multiple parents. |

#### Can They Coexist? Yes! 🤝
A class hierarchy can combine both:

```python
class Human: pass
class Parent: pass

class Male(Parent, Human):  # 1. Multiple Inheritance (Parent + Human -> Male)
    pass

class Boy(Male):            # 2. Multilevel Inheritance (Male -> Boy)
    pass
```
* **Hierarchy:** `(Parent, Human) ➔ Male ➔ Boy`

````python
class Human:
    can_breath=True # class variable
    def __init__(self,num_heart):
        print("calling init from male class")
        self.eyes=2
        self.heart=num_heart
    def eat(self):
        print("I can eat")
    def work(self):
        print("I can work")
class Male(Human):
    def __init__(self,name):
        print("calling init from male class")
        self.name=name
    def sleep(self):
        print("I can sleep")
class Boy(Male):
    def __init__(self, can_dance,heart,name):
        Human.__init__(self,heart)
        Male.__init__(self,name)
        self.know_dancing=can_dance
    def  work(self):
        super().work()
        print("I can code")

boy_1=Boy('yes',1,'yeshu')
print(boy_1.name)
print(boy_1.know_dancing)
print(boy_1.can_breath)
# print(Boy.mro())
# print(type(Human))
````
