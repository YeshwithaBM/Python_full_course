### Abstraction

- **Abstraction** means **showing only the essential functionality to the user while hiding the internal implementation details**.

- It focuses on **what an object does**, rather than **how it does it**.

- Abstraction helps us **hide unnecessary complexity** and expose only the important features.

-  **abstraction means exposing only the necessary interface/functionality while hiding unnecessary implementation details**.

- Abstraction is about hiding unnecessary complexity and exposing essential functionality. Encapsulation is one of the mechanisms that can help achieve this by bundling and controlling access to data and implementation.
### Abstract Class and `ABC` Module

- Python provides the **`abc` module** (`abc` = **Abstract Base Classes**) to create and work with **abstract classes**.

- We can import `ABC` from the `abc` module:

```python
from abc import ABC, abstractmethod

# Abstraction in Python

- **Abstraction** means showing **what an object can do** while hiding the **unnecessary details of how it does it**.
- Example: `upi.pay()` exposes the payment operation, while the internal steps like validation, balance checking, sending the request, and confirmation are hidden inside `pay()`.

**WHAT it does → exposed**  
**HOW it does it → hidden**

## Abstract Class

- An **abstract class** is a **blueprint/contract** for its child classes.
- It defines functionality that child classes **must implement**.
- In Python, we use `ABC` from the `abc` module.

    from abc import ABC, abstractmethod

    class Payment(ABC):

        @abstractmethod
        def pay(self):
            pass

- `Payment` → Abstract class
- `pay()` → Abstract method
- `ABC` → used to create an Abstract Base Class

## Abstract Method

- An **abstract method** is a method declared in an abstract class that the child class **must implement**.

    @abstractmethod
    def pay(self):
        pass

- The parent defines **WHAT is required**.
- The child decides **HOW to implement it**.

    class CreditCard(Payment):
        def pay(self):
            print("Payment using Credit Card")

    class UPI(Payment):
        def pay(self):
            print("Payment using UPI")

- Both classes must have `pay()`, but each can implement it differently.

## Why Do We Need an Abstract Method?

    Payment
       |
    -------------
    |           |
    CreditCard  UPI
       |         |
      pay()     pay()

- Every payment type should have a `pay()` operation.
- But the implementation can be different for each payment type.
- Therefore, the parent defines the **common requirement**, and the child provides the **actual implementation**.

## What Happens If We Create an Object of the Abstract Class?

    p = Payment()

- This gives a `TypeError`.
- An abstract class containing unimplemented abstract methods **cannot be instantiated**.

    Payment
       ↓
    abstract pay()
       ↓
    cannot create its object

## What If the Child Does Not Implement the Abstract Method?

    class CreditCard(Payment):
        pass

    card = CreditCard()

- This gives a `TypeError`.
- `CreditCard` is still considered **abstract** because it has not implemented `pay()`.

- After implementing it:

    class CreditCard(Payment):
        def pay(self):
            print("Payment using Credit Card")

    card = CreditCard()

- Now the object can be created.

## Where Is the "Hiding" in Abstraction?

- Suppose UPI internally performs several steps:

    class UPI(Payment):

        def pay(self):
            self.validate_upi()
            self.check_balance()
            self.send_request()
            self.confirm_payment()

- The user only needs to call:

    upi.pay()

- The user does not need to know or manually call:

    validate_upi()
    check_balance()
    send_request()
    confirm_payment()

- Therefore, the **complex internal implementation is hidden behind the simple `pay()` interface**.

**Abstraction → WHAT the object can do**  
**Implementation → HOW the object does it**

## What Is `@abstractmethod`?

- `@abstractmethod` is a **decorator**.
- It tells Python that the method is **abstract** and that concrete child classes must implement it.

    @abstractmethod
    def pay(self):
        pass

## What If We Write Code Inside an Abstract Method?

- An abstract method does **not** have to contain only `pass`.

    @abstractmethod
    def pay(self):
        print("Payment started")

- It is still abstract because it has `@abstractmethod`.
- `pass` is commonly used because the parent is usually only defining the **contract**, while the child provides the actual implementation.

## Important Terms

- **Abstraction** → showing WHAT and hiding unnecessary HOW
- **Abstract Class** → blueprint/contract for child classes
- **Abstract Method** → method that the child class must implement
- **ABC** → used to create an Abstract Base Class
- **`@abstractmethod`** → decorator that marks a method as abstract
- **Implementation** → the actual HOW of performing the operation

````python
from abc import ABC,abstractmethod
class Payment(ABC):
    @abstractmethod
    def pay():
        pass
# p=Payment()
#   ^^^^^^^^^
# TypeError: Can't instantiate abstract class Payment without an implementation for abstract method 'pay'
#"This class is only a blueprint/interface. Don't create objects directly from it, and subclasses MUST implement this method."
#That's an abstract class.
#Now Payment is an abstract class.
#And pay() is an abstract method.
````

````python

class CreditCard(Payment):
    def __init__(self):
        self.balance=1600

    def pay(self):
        print("Payment using Credit Card")

class UPI(Payment):
    def __init__(self,money):
        self.balance=1600
        self.money=money
    def pay(self):

        self.validate_upi()
        self.check_balance()

    def validate_upi(self):
        print("Validating UPI...")

    def check_balance(self):
        if self.money>self.balance :
            print("Insufficient Balance")
        
        else:
            self.send_request()
            self.confirm_payment()
            print("Payment successful using UPI")
        
 
    def send_request(self):
        print("Sending payment request...")

    def confirm_payment(self):
        print("Confirming payment...")
````

````python
card = CreditCard()
upi = UPI(1500)

# card.pay()
upi.pay()
````
