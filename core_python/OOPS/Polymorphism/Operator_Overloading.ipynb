# Operator Overloading

**Operator overloading** means defining custom behavior for an existing operator when it is used with objects of our class.

### Why?
- Makes objects easier and more natural to use.
- Allows the same operator to behave differently for different objects.
- Supports **polymorphism**.

### How?
Python uses **special/dunder methods**:

    +   → __add__()
    -   → __sub__()
    *   → __mul__()
    >   → __gt__()
    <   → __lt__()
    ==  → __eq__()

Example:

    class Person:
        def __init__(self, age):
            self.age = age

        def __gt__(self, other):
            return self.age > other.age

    p1 = Person(30)
    p2 = Person(25)

    print(p1 > p2)       # True

`p1 > p2` internally uses:

    p1.__gt__(p2)

### Operator Overloading & Polymorphism

The same operator can behave differently depending on the objects:

    1 + 2          → addition
    "1" + "2"      → concatenation
    c1 + c2        → custom ComplexNumber addition

So:

> **Same operator → different behavior for different objects = polymorphism.**

### One-line definition

> **Operator overloading is defining how an existing operator should behave with objects of a user-defined class.**

```python
print(1+2) # here, the + operator is used to add two integers, resulting in the integer 3.
print('1' + '2') # here, the + operator is used to concatenate two strings, resulting in the string '12'.
```

```python
class ComplexNumber:
    def __init__(self,r,i):
        self.real = r
        self.imaginary = i
    def __add__(self, other):
        return f"{self.real + other.real} + {self.imaginary + other.imaginary}i"
c1=ComplexNumber(1,2)
c2=ComplexNumber(3,4) 
print(c1 + c2) # here, the + operator is overloaded to add two ComplexNumber objects, resulting in the string '4 + 6i'.  
print(c1.__add__(c2)) # here, the __add__ method is called directly to add two ComplexNumber objects, resulting in the same string '4 + 6i'.
# this + operator is performing beyond its default behavior of adding numbers or concatenating strings, and is instead performing a custom operation defined in the ComplexNumber class.
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def __gt__(self, other):
        if self.age > other.age :
            return True
        else:
            return False

p1=Person("mohan", 30)
p2=Person("yeshu", 25)
print(p1.__gt__(p2)) # here, the __gt__ method is called directly to compare the ages of two Person objects, resulting in the True boolean value.
print(p1 > p2) # here, the > operator is overloaded to compare the ages of two Person objects, resulting in the same True boolean value.
if p1 > p2:
    print(f"{p1.name} is paying the bill")
else:
    print(f"{p2.name} is paying the bill")
```
