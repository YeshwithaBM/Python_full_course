# Polymorphism

**Poly** → Many  
**Morphism** → Forms

**Polymorphism** means the same operation, function, or interface can behave differently depending on the type of data or object.

## Example: `+` Operator

`+` behaves differently depending on the data type:

    10 + 20              # 30
    "Hello " + "World"   # "Hello World"

- Integer + Integer → **Addition**
- String + String → **Concatenation**

## Example: `len()` Function

The same `len()` function works differently with different data types:

    len("Python")              # 6
    len([10, 20, 30])          # 3
    len({"name": "Alex"})      # 1

- String → Number of characters
- List → Number of elements
- Dictionary → Number of key-value pairs

## Why Is Polymorphism Useful?

Polymorphism allows us to write **general and reusable code** without manually checking the type of data.

    def display_length(data):
        print(len(data))

    display_length("Python")
    display_length([10, 20, 30])
    display_length({"name": "Alex"})

The same function works with different types because `len()` automatically performs the appropriate operation for each type.

## Simple Definition

> **Polymorphism is the ability of the same operation, function, or interface to perform different behavior depending on the type of data or object.**

## Key Idea

**Same operation/interface → Different behavior**

```python
def display_length(data):
    print(len(data))

display_length("Python")
display_length([10, 20, 30])
display_length({"name": "Alex"})
```

```python
num1=29
num2=3
print(num1 + num2) # addition of numbers

a = "Hello"
b = "World"

print(a + b) # concatenation of strings
```

> **"What is the ONE thing that stays the same?"**

| Feature | What stays the same? | What changes? |
|---|---|---|
| Method Overriding | Method name | Implementation |
| Method Overloading | Method name | Parameters/behavior |
| Operator Overloading | Operator | Operation/behavior |
| Duck Typing | Expected method/behavior | Object providing it |

Therefore, they all connect to **polymorphism** because:

> **Something common is used in multiple forms, producing different behavior.**

### The simplest mental model

    POLYMORPHISM
         |
         |--- same method → different behavior
         |       ↓
         |   overriding
         |
         |--- same method → different parameters
         |       ↓
         |   overloading
         |
         |--- same operator → different behavior
         |       ↓
         |   operator overloading
         |
         |--- same expected behavior → different objects
                 ↓
             duck typing

## One sentence to remember

> **Polymorphism is the main concept; overriding, overloading, operator overloading, and duck typing are different ways of achieving or expressing “one common interface/operation with multiple forms of behavior.”**
