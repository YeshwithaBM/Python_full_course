# Polymorphism

**Poly** → Many  
**Morphism** → Forms

**Polymorphism** means the same operation, function, or interface can behave differently depending on the type of data or object.

## Example: `+` Operator

`+` behaves differently depending on the data type:

    10 + 20              # 30
    "Hello " + "World"   # "Hello World"

- Integer + Integer → **Addition**
- String + String → **Concatenation**

## Example: `len()` Function

The same `len()` function works differently with different data types:

    len("Python")              # 6
    len([10, 20, 30])          # 3
    len({"name": "Alex"})      # 1

- String → Number of characters
- List → Number of elements
- Dictionary → Number of key-value pairs

## Why Is Polymorphism Useful?

Polymorphism allows us to write **general and reusable code** without manually checking the type of data.

    def display_length(data):
        print(len(data))

    display_length("Python")
    display_length([10, 20, 30])
    display_length({"name": "Alex"})

The same function works with different types because `len()` automatically performs the appropriate operation for each type.

## Simple Definition

> **Polymorphism is the ability of the same operation, function, or interface to perform different behavior depending on the type of data or object.**

## Key Idea

**Same operation/interface → Different behavior**

```python
def display_length(data):
    print(len(data))

display_length("Python")
display_length([10, 20, 30])
display_length({"name": "Alex"})
```
