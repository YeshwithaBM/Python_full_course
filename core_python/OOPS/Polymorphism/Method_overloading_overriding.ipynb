#### Method Overloading

**Method overloading** means defining multiple methods with the **same name but different parameters within the same class**.

- Python does **not support traditional method overloading**.

If we define the same method multiple times:

    class Calculator:
        def add(self, a):
            print(a)

        def add(self, a, b):
            print(a + b)

The **latest definition replaces the previous one**.

So Python keeps only:

    add(self, a, b)

### How can we achieve similar behavior?

We can use:

1. **Default arguments**

       def add(self, a, b=0):
           return a + b

2. **`*args`**

       def add(self, *numbers):
           return sum(numbers)

So Python achieves **method-overloading-like behavior** using flexible arguments rather than traditional method overloading.

---

```python
class Demo:
    def add(self,a,b):# here this method is discarded and replaced by the next method with the same name
        return a+b
    def add(self,a,b,c):
        return a+b+c
d=Demo()
print(d.add(10,20)) # this will give an error because the method with two parameters is discarded   
print(d.add(10,20,30))
# but we can use default parameters to achieve the method overloading in python
```

```python
# here we are implementing method overloading using default parameters, so we can call the method with two or three parameters
class Demo:
    def add(self,a,b,c=0): # here we are using default parameter c=0, so if we call the method with two parameters then c will take the default value of 0
        return a+b+c

d=Demo()
print(d.add(10,20)) # this will work fine because the default value of c        
print(d.add(10,20,30))
```

```python
# another way to implement method overloading is by using variable length arguments, 
# so we can call the method with any number of parameters
class Demo:
    def add(self,*args): # here we are using variable length arguments, so we can call the method with any number of parameters
        total=0
        for i in args:
            total+=i
        return total
d=Demo()
print(d.add(10,20)) # this will work fine because we are using variable length          
print(d.add(10,20,30))
print(d.add(10,20,30,40,50))
```

### Method Overriding

**Method overriding** happens when a **child class provides its own implementation of a method that already exists in the parent class**.

It requires:

- At least **two classes**
- **Inheritance**
- Same method name
- Child provides a new implementation

    class Animal:
        def speak(self):
            print("Animal speaks")


    class Dog(Animal):
        def speak(self):
            print("Dog barks")


    dog = Dog()
    dog.speak()

    #### Dog barks

Python first looks for `speak()` in the **child class**.

If it exists → the child's method is used.

If it doesn't exist → Python continues searching through the **parent classes according to the MRO**.


```python
class Father:
    def sleep(self):
        print("sleeps from 10pm to 6am")
    def eat(self):
        print("eats 3 times a day")

class Son(Father):
    pass
    # def sleep(self):
    #     print("sleeps from 12am to 8am")
    # def eat(self):
    #     print("eats 2 times a day")
ram=Son()
ram.sleep()  #here the sleep method is called since the sleep method is defined in father class and  he inherits the sleep method from father class, 
#but if we define the sleep method in son class then the sleep method of son class will be called instead of father class,
#  this is called method overriding
```

```python
class Father:
    def sleep(self):
        print("sleeps from 10pm to 6am")
    def eat(self):
        print("eats 3 times a day")

class Son(Father):
    # pass
    def sleep(self):
        print("sleeps from 12am to 8am")
        # super().sleep() # here we are calling the sleep method of father class using super() function, so the output will be "sleeps from 12am to 8am" and "sleeps from 10pm to 6am"
    def eat(self):
        print("eats 2 times a day")
ram=Son()
ram.sleep() # here the sleep method of son class is called instead of father class, this is called method overriding 
```

### Key Difference

| Method Overloading | Method Overriding |
|---|---|
| Same class | Parent + child classes |
| Same method name | Same method name |
| Different parameters | Child provides new implementation |
| Python doesn't support traditional overloading | Python supports overriding |
| Can simulate using `*args` / default arguments | Uses inheritance |

### Easy memory trick

**Overloading → same class, same name, different inputs**

**Overriding → child class replaces parent's behavior**
- Overloading → Compile-time polymorphism
- Overriding → Runtime polymorphism
