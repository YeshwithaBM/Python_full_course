### Duck Typing in Python

#### 1. What is Duck Typing?

**Duck typing** is a programming approach where Python cares about **what an object can do**, rather than **what type or class the object belongs to**.

> **“If it walks like a duck and quacks like a duck, treat it like a duck.”**

In simple words:

- We don't ask: **“What class is this object?”**
- We ask: **“Does this object have the methods/behavior I need?”**

---

#### 2. Basic Example

    class Duck:
        def swim(self):
            print("Duck is swimming")

        def speak(self):
            print("Quack")


    class Dog:
        def swim(self):
            print("Dog is swimming")

        def speak(self):
            print("Woof")


    def display(animal):
        animal.swim()
        animal.speak()


    display(Duck())
    display(Dog())

### Output

    Duck is swimming
    Quack

    Dog is swimming
    Woof

Notice that `Duck` and `Dog` do **not** inherit from a common parent class.

Still, `display()` works with both because both objects provide:

    swim()
    speak()

This is **duck typing**.

---

# 3. What Exactly Happens?

When we write:

    display(Dog())

Python receives a `Dog` object.

Inside `display()`:

    animal.swim()

Python effectively asks:

> “Does this object have a `swim()` method?”

`Dog` has it, so Python calls it.

Then:

    animal.speak()

Python asks:

> “Does this object have a `speak()` method?”

`Dog` has it, so Python calls it.

Python does **not** need to check:

    isinstance(animal, Duck)

It only cares about the required behavior.

---

# 4. What If the Required Method Does Not Exist?

Suppose we have:

    class Car:
        def drive(self):
            print("Car is driving")


    display(Car())

`display()` expects:

    animal.swim()
    animal.speak()

But `Car` does not have `swim()`.

Therefore Python raises:

    AttributeError

This shows an important point:

> Duck typing is generally checked **at runtime** when the required operation is performed.

Python does not require every possible method to be declared beforehand.

---

# 5. Why is Duck Typing Related to Polymorphism?

### Polymorphism

**Polymorphism = one interface/operation, many forms of behavior.**

For example:

    def display(animal):
        animal.swim()
        animal.speak()

The same function:

    display()

can work with different objects:

    display(Duck())
    display(Dog())

But each object behaves differently.

Duck:

    swim() → Duck swimming
    speak() → Quack

Dog:

    swim() → Dog swimming
    speak() → Woof

So:

> **Same interface → different behavior**

That is polymorphism.

**Duck typing is one way Python achieves polymorphic behavior.**

---

# 6. Does Duck Typing Require Inheritance?

**No.**

This is one of its important features.

For example:

    class Duck:
        def speak(self):
            print("Quack")


    class Dog:
        def speak(self):
            print("Woof")


    def make_sound(animal):
        animal.speak()

There is no inheritance relationship between `Duck` and `Dog`.

Still:

    make_sound(Duck())
    make_sound(Dog())

works.

The function only cares that the object has:

    speak()

---

# 7. Why Do We Need Duck Typing?

The main purpose is:

> **To make code flexible and reusable.**

Without duck typing, we might unnecessarily force unrelated classes to inherit from the same parent just so a function can accept them.

With duck typing:

    def make_sound(animal):
        animal.speak()

Any object that provides `speak()` can be used.

For example:

    class Dog:
        def speak(self):
            print("Woof")


    class Cat:
        def speak(self):
            print("Meow")


    class Robot:
        def speak(self):
            print("Beep")


    def make_sound(obj):
        obj.speak()

All three can work:

    make_sound(Dog())
    make_sound(Cat())
    make_sound(Robot())

The function does not care whether the object is a `Dog`, `Cat`, or `Robot`.

It only cares about the required behavior:

    speak()

---

# 8. Real-World Example

A very common example is a function that writes data:

    def save_data(file):
        file.write("Hello")

The function does not necessarily need to know the exact class of `file`.

It only needs an object that provides:

    write()

So different objects that support `write()` can potentially be used.

The important idea is:

> **Program to the behavior you need, not the exact type of object.**

---

# 9. Duck Typing vs Dynamic Typing

These concepts are related but **not the same**.

### Dynamic Typing

Python determines the type of a variable at runtime.

Example:

    x = 10
    x = "Hello"
    x = [1, 2, 3]

The same variable name can refer to objects of different types.

### Duck Typing

Python focuses on whether an object supports the required operation.

Example:

    def square(x):
        return x * x

The function does not explicitly specify:

    x must be an int

It simply performs:

    x * x

So `x` needs to support multiplication.

### Key Difference

**Dynamic typing:**

> “The type is determined at runtime.”

**Duck typing:**

> “I care about what operations the object supports, not its exact type.”

---

# 10. Duck Typing and Inheritance

### Traditional inheritance-based approach

You might create:

    class Animal:
        def speak(self):
            pass


    class Dog(Animal):
        def speak(self):
            print("Woof")


    class Cat(Animal):
        def speak(self):
            print("Meow")

Here `Dog` and `Cat` share a common parent.

### Duck typing approach

You can simply write:

    class Dog:
        def speak(self):
            print("Woof")


    class Cat:
        def speak(self):
            print("Meow")


    def make_sound(animal):
        animal.speak()

No common parent is required.

Both approaches can provide polymorphic behavior.

---

# 11. How is Duck Typing Written?

There is **no special `duck` keyword** and no special syntax.

You simply write code that operates on the required behavior.

Example:

    def process(obj):
        obj.run()

Any object with a compatible `run()` method can potentially be passed.

The function does not need to explicitly check:

    if isinstance(obj, SomeClass):

That is the basic idea of duck typing.

---

# 12. Important Point

Duck typing does **not** mean:

> “Python checks whether the object is a duck.”

It means:

> “Python attempts to use the object according to the behavior we need.”

For example:

    obj.speak()

Python attempts to call `speak()`.

If the object supports it → it works.

If it doesn't → Python raises an appropriate runtime error such as:

    AttributeError

---

# 13. Why is Duck Typing Useful?

Duck typing provides:

- **Flexibility** — different object types can be used.
- **Reusability** — the same function can work with many objects.
- **Less unnecessary inheritance** — unrelated classes don't need a common parent.
- **Polymorphism** — different objects can respond to the same operation differently.
- **Simple interfaces** — code focuses on required behavior rather than exact types.

---

```python
class Duck:
    def swim(self):
        print("Duck is swimming")
    def speak(self):
        print("Duck says Quack!")
class Dog:
    def swim(self):
        print("Dog is swimming")
    def speak(self):
        print("Dog says Woof!")

def display(d):
    d.swim()
    d.speak()
    print("Information about the animal is displayed above.")
d = Duck()
dog = Dog()
display(dog)
```

```python
#  your main goal is to create a function that can send notifications through different services (Email, SMS, WhatsApp) without worrying about the specific implementation of each service.
#  This is a classic example of duck typing in Python, where the focus is on the behavior (methods) of the objects rather than their specific types.
class Email:
    def send(self, message):
        print(f"Sending Email: {message}")


class SMS:
    def send(self, message):
        print(f"Sending SMS: {message}")


class WhatsApp:
    def send(self, message):
        print(f"Sending WhatsApp message: {message}")


def notify(service, message):
    service.send(message)


email = Email()
sms = SMS()
whatsapp = WhatsApp()

notify(email, "Your order has been shipped.")
notify(sms, "Your order has been shipped.")
notify(whatsapp, "Your order has been shipped.")
```
