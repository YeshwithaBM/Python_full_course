## Hierarchical Inheritance

- Multiple child/subclass/derived classes inherit from **one parent/superclass/base class**.

- This is called **hierarchical inheritance** because multiple child classes inherit from the same single parent class.

- Multiple child classes created from the same parent will **not have any inheritance relationship with each other**. They only have an inheritance relationship with the common parent class.

```text
              Parent
             /      \
            ↓        ↓
         Child 1   Child 2

````python
class Human:
    def __init__(self,name,age):
        print("Human init")
        self.name=name
        self.age=age
    def showDetails(self):
        print(f"name: {self.name}, age: {self.age}")

    def eat(self):
        print("I can eat")

class Male(Human):
    def __init__(self, name, age,location):
        super().__init__(name, age)
        self.location=location
    def showDetails(self):
        super().showDetails()
        print(f'location: {self.location}')
    def sleep(self):
        print("I can sleep")

class Female(Human):
    def __init__(self, name, age,can_dance):
        super().__init__(name, age)
        self.can_dance=can_dance
    def showDetails(self):
        super().showDetails()
        print(f'know dancing: {self.can_dance}')
    def work(self):
        print("I can code ")

female_1=Female("yeshu",20,"Yes")
# print(female_1.can_dance)
# print(female_1.name)
female_1.showDetails()
male_1=Male("Mohan",23,"bangalore")
male_1.showDetails()
# print(male_1.name)
# print(Male.mro())
````

#### Key points to remember
- Hierarchical inheritance → multiple child classes inherit from one parent class.
- Male and Female are siblings because they inherit from the same parent Human.
- Common attributes/methods can be placed in Human and reused by both children.
- Each child can have its own additional attributes and methods.
- Each child can override a parent's method.
- super() allows the child to reuse the parent's implementation.
- The children (Male and Female) are separate classes; Male does not inherit anything from Female, and vice versa.

### Hybrid Inheritence 
- combination of two or more inheritence .

````python
class A:
    def display(self):
        print("display from A class")
class B(A):
    def display(self):
        print("display from B class")
class C:
    def display(self):
        B.display(self)
        print("display from c class")
class D(B,C):
    def display(self):
        print("display from D class")
d1=D()
d1.display()
print(D.mro())
c1=C()
c1.display()
````

````python
class University:
    def __init__(self,uni_name):
        self.university=uni_name
    def showDetails(self):
        print(f"unversity is {self.university}")
class Course(University):
    def __init__(self,course_name,uni_name):
        University.__init__(self,uni_name)
        self.course=course_name
    def showDetails(self):
        super().showDetails()
        print(f"course is {self.course}")
class Branch(University):
    def __init__(self,branch_name,uni_name):
        University.__init__(self,uni_name)
        self.branch=branch_name
    def showDetails(self):
        super().showDetails()
        print(f"branch is {self.branch}")

class Student(Course,Branch):
    def __init__(self, student_name,course,branch,uni):
        Course.__init__(self,course,uni)
        Branch.__init__(self,branch,uni)
        self.student=student_name
    def showDetails(self):
        Course.showDetails(self)
        Branch.showDetails(self)
        print(f'Student name is {self.student}')

class Faculty(Branch):
    def __init__(self, faculty_name,branch,uni):
        Branch.__init__(self,branch,uni)
        self.faculty=faculty_name
    def showDetails(self):
        super().showDetails()
        print(f'Faculty name is {self.faculty}')

s=Student("Yeshu","Btech","AIML","SU")
s.showDetails()
        

````

````python
class University:
    def __init__(self, uni_name):
        self.university = uni_name

    def showDetails(self):
        print(f"University is {self.university}")


class Course(University):
    def __init__(self, course_name, uni_name):
        University.__init__(self, uni_name)
        self.course = course_name

    def showDetails(self):
        super().showDetails()
        print(f"Course is {self.course}")


class Branch(University):
    def __init__(self, branch_name, uni_name):
        University.__init__(self, uni_name)
        self.branch = branch_name

    def showDetails(self):
        super().showDetails()
        print(f"Branch is {self.branch}")


class Student(Course, Branch):
    def __init__(self, student_name, course, branch, uni):

        # Initialize Course part
        Course.__init__(self, course, uni)

        # Initialize Branch part
        Branch.__init__(self, branch, uni)

        # Student's own attribute
        self.student = student_name

    def showDetails(self):
        print(f"University is {self.university}")
        print(f"Course is {self.course}")
        print(f"Branch is {self.branch}")
        print(f"Student name is {self.student}")


class Faculty(Branch):
    def __init__(self, faculty_name, branch, uni):

        # Initialize Branch part
        Branch.__init__(self, branch, uni)

        # Faculty's own attribute
        self.faculty = faculty_name

    def showDetails(self):
        super().showDetails()
        print(f"Faculty name is {self.faculty}")


s = Student("Yeshu", "Btech", "AIML", "SU")
s.showDetails()
````
