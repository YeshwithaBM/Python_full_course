## Access Specifiers in Python

Python commonly uses **three access levels**:

1. **Public**
2. **Protected**
3. **Private**

### 1. Public

- Public members can be accessed **from anywhere**.
- They can be accessed:
  - Inside the same class
  - Inside derived/child classes
  - Outside the class

### 2. Protected
- Protected members are intended to be accessed:
    - Inside the class
    - Inside derived/child classes
- They are not strictly enforced as private in Python.
- Python uses _ as a convention to indicate that a member is intended for internal/derived-class use.

### 3. Private
- Private members are intended to belong only to the class that defines them.
- They are not directly accessible using the normal attribute name from:
    - Outside the class
    - A derived/child class
- Python implements this using name mangling.

**Important: Python does not have Java/C++-style strict access modifiers. public, protected, and private in Python are mainly implemented through naming conventions and name mangling.**

````python
#Public 
class Student:
    def __init__(self,name):
        self.name=name # by default it is public 
    def display(self):
        print(f"Hi myself {self.name} from student class ")
# by default attributes and methods are public 
s1=Student("yeshu")
print(s1.name)
s1.display()
````

````python
# Protected
'''So the key point
_ does NOT provide protection.
It simply tells other programmers:
"This member is intended to be internal/protected; don't treat it as part of the public interface."
That's why Python's protected access is often described as a convention rather than enforcement'''

class Student:
    def __init__(self,name,rollno):
        self.name=name  # Public instance variable 
        self._rollno=rollno # protected instance variable 
    def display(self):
        print(f"Hi myself {self.name} with {self._rollno} rollno from student class ")
class Branch(Student):
    pass

# def show():
#     b1= Branch("yeshu", 23)
#     print(b1.name)
#     print(b1._rollno)

# show()
s1=Student("Mohan",29)
s1.name="yeshu"
s1._rollno=34
print(s1.name)
s1.display()
````

````python
# Private 

class Student:
    def __init__(self,name,rollno,age):
        self.name=name  # Public instance variable 
        self._rollno=rollno # protected instance variable 
        self.__age=age
    def __display(self):
        print(f"Hi myself {self.name} {self.__age} yrs old with {self._rollno} rollno from student class ")
    def displayPrivatedata(self):
        self.__display()


class Branch(Student):
    def show(self):
        print(f"my age is {self.__age}")#cannot be access since its private 
b1=Branch("nish",23,26)

print(b1.__age)  # cannot be access since its private 
b1.show()

s1=Student("Mohan",29,20)
s1.displayPrivatedata() # can be access through function since its inside that class and  we are not actually accessing outside 
# cannot be access since its private 
# but python doesnt have strict restrictions , we still can access using other methods called name mangling 
print(s1.__age)
s1.display()  # cannot be access since its private 
````

**name mangling is the process in Python where an attribute or method beginning with two underscores (__) is automatically renamed internally by adding the class name to its name, mainly to prevent accidental access or name conflicts.**
