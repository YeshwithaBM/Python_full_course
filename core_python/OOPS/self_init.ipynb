### Constructor

#### `self` and `__init__`
- Basically, the task of a **constructor** is to initialize the data members (attributes) of a class by assigning values to them when an object is created.
- `__init__` is commonly called the **constructor** in Python.
- It is a **special method** that is automatically called when an object is created.
- Its main purpose is to **initialize the attributes of the object**.
- `__init__` is a **special method** in Python that is automatically called when an object is created.
- The `__init__` method is used to **initialize the attributes of an object**.
- Python recognizes `__init__` as a special method because of its predefined name.
- `self` is a **parameter** used inside instance methods, including `__init__`.
- `self` refers to the **actual object (instance)** on which the method is being called.
- `self` is **not a keyword** in Python. It is a conventional parameter name. We normally use `self`, but technically another name can be used.
- When we create an object, Python automatically calls the `__init__` method and passes the newly created object as the `self` argument.
#### Example
```python
class Instructor:
    def __init__(self,name,address): 
        self.name=name
        self.address=address

instructor_1 = Instructor("yeshu","banglore")
print(instructor_1.name)

# When we write:
instructor_1 = Instructor() 
Python effectively passes the newly created object as the first argument to __init__:
self → instructor_1 # first parameter 
name → "yeshu" # next
address → "banglore"
