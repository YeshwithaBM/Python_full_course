```python
class Circle:
    pi=3.14
    def __init__(self,radius):
        self.radius=radius
    def area(self):
        area_of_circle=self.pi*(self.radius**2)
        print(f'The area of circle is {area_of_circle}')
    def circumference(self):
        circumference_of_circle=2*self.pi*self.radius
        print(f"The circumference of circle is {circumference_of_circle}")
circle_1=Circle(6)
circle_1.area()
circle_1.circumference()
```
