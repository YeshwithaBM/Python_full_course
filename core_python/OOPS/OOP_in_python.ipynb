### OOP in python 

## POP — Procedural-Oriented Programming

- Basically, without using the OOP concept, we can follow the **POP approach**.
- POP = **Procedural-Oriented Programming**.
- In POP, the program is mainly organized around **procedures/functions**.
- A **procedure** means a sequence of steps/instructions used to perform a particular task.
- Main focus of POP is:
  - **What tasks need to be performed?**
  - **What steps are required to perform those tasks?**
- Data and functions are generally handled separately.

### Why do we need OOP?

- POP works well for **small and straightforward programs**.
- But as the project becomes larger, managing many functions and shared data can become difficult.
- In POP, multiple functions may access and modify the same data, especially when using **global variables**.
- This can make it harder to control and protect data.
- There is less organization between **data and the operations that work on that data**.
- As the number of tasks/functions increases, the program can become more difficult to understand, maintain, and modify.
- Changes in one part of a large procedural program can sometimes affect other functions that depend on the same data.
- Therefore, for large and complex programs, we need a better way to organize **data + related operations**.
- This is where **OOP** becomes useful.

### Main Idea of OOP

- OOP organizes a program around **objects**.
- An object groups:
  - **Data** → attributes
  - **Functions that operate on the data** → methods

```text
POP
│
├── Data
│
└── Functions → perform tasks


OOP
│
└── Object
     ├── Data
     └── Methods → operate on that data


## Why is Everything in Python an Object?

- Python was designed around the idea that **values should be treated uniformly**.
- Instead of having separate rules for primitive values, functions, classes, etc., Python represents them all as **objects**.
- This makes the language **consistent and flexible**.

### Why is this useful?

#### 1. Same model for different things

- In some languages, basic values like integers are treated differently from objects.
- Python avoids this distinction.
- An integer, string, list, function, etc. can all be treated using the same general object model.

#### 2. Everything can have behavior

- Objects can have operations associated with them.
- For example, a string has operations like `.upper()`, a list has `.append()`, etc.
- This keeps **data and the operations related to that data** together.

#### 3. Functions can be treated like data

- Because functions are objects, they can be:
  - Stored in variables
  - Passed to other functions
  - Returned from functions
  - Stored in lists/dictionaries

This makes Python more flexible and supports **functional programming** as well.

#### 4. Classes can also be treated as objects

- Python even treats classes as objects.
- This allows Python to have a very flexible and dynamic object system.

#### 5. It keeps Python consistent

Instead of having completely different rules for:

```text
integer
string
list
function
class

Python's special design choice is not that other languages don't have functions or methods. It's that Python represents almost all runtime values using one common object model.

### Object Model

- Object model = the rules a programming language uses to represent and work with objects.
- Python has a unified object model, where almost everything is treated as an object.
- This gives different types of values a common way of being handled.
