# Inheritance

### What is Inheritance and Why is it Needed?

- **Inheritance** is the ability of a newly created class to **inherit attributes and methods from an already existing class**.

- In other words, we can create a new class based on an existing class and reuse the existing class's **attributes and methods**.

- The class from which we inherit the attributes and methods is called the:
  - **Base class**
  - **Parent class**
  - **Superclass**

- The class that inherits the attributes and methods from the parent class is called the:
  - **Derived class**
  - **Child class**
  - **Subclass**

### Syntax

```python
class DerivedClassName(BaseClassName):
    # class definition
```
#### Why do we need Inheritance?
- Code reusability — we can reuse attributes and methods that are already defined in the parent class.
- Avoids code duplication — we don't have to write the same attributes and methods again in the child class.
- Extends existing functionality — the child class can use the parent's features and can also have its own additional attributes and methods.

````python
class Human:
    def eat(self):
        print("I can eat")
    def work(self):
        print("I can work")
class Male:
    def eat(self):
        print("I can eat")
    def work(self):
        print("I can work")
male_1=Male()
male_1.eat()
````

````python
class Human:
    def __init__(self,num_heart):
        self.num_eyes=2
        self.num_nose=1
        self.num_heart=num_heart
    def eat(self):
        print("I can eat")
    def work(self): # Inheritence 
        print("I can work")
class Male(Human):
    def __init__(self,name,heart):
        self.name=name
        super().__init__(heart)
    def flirt(self):
        print("I can flirt")
    def work(self):
        super().work() 
        print("I can code")#method overriding
    def display(self):
        print(f"Hi, I'm {self.name} and I have {self.num_heart} heart. ")
male_1=Male("yeshu",1)
# male_1.flirt()
# male_1.eat()
# male_1.work()
print(male_1.name)
print(male_1.num_eyes) 
print(male_1.num_heart)
male_1.display()
````

### Types of Inheritence 
- single level Inheritence 
- multiple inheritence 
- multi-level inheritence 
- heirarchical inheritence 
- hybrid inheritence 

#### Single-Level Inheritance

- One child/subclass/derived class inherits from **one parent/superclass/base class**.

````python
class Human:
    def __init__(self,num_heart):
        self.num_eyes=2
        self.num_nose=1
        self.num_heart=num_heart
    def eat(self):
        print("I can eat")
    def work(self): # Inheritence 
        print("I can work")
class Male(Human):
    def __init__(self,name,heart):
        self.name=name
        super().__init__(heart)
    def flirt(self):
        print("I can flirt")
    def work(self):
        super().work() 
        print("I can code")#method overriding
    def display(self):
        print(f"Hi, I'm {self.name} and I have {self.num_heart} heart. ")
male_1=Male("yeshu",1)
# male_1.flirt()
# male_1.eat()
# male_1.work()
print(male_1.name)
print(male_1.num_eyes) 
print(male_1.num_heart)
male_1.display()
````

#### Multiple Inheritence .
- One child/subclass/derived class inherits from **more than one parent/superclass/base class**.

````python
class Human:
    def eat(self):
        print("I can eat")

    def work(self):
        print("I can work")

class Male:
    def flirt(self):
        print("I can flirt")

    def work(self):
        print("I can code")

class Boy(Human, Male):  # Order matters if methods have the same name
    def sleep(self):
        print("I can sleep")

    def work(self):
        print("I can test")

boy_1 = Boy()

boy_1.eat()
boy_1.flirt()
boy_1.work()
# First searches in its own class,
# then searches the parent classes in the order specified
# in the class definition: Human, then Male.

Male.work(boy_1)
# If I want to specifically access the method from Male,
# I can call it directly like this.

print(Boy.mro())
# Shows the Method Resolution Order (MRO),
# which is the order Python follows when searching for a method.
````

````python
class Human:
    def __init__(self, num_heart):
        print("calling human init function")

        # Attributes belonging to the Human part of the object
        self.num_eyes = 2
        self.num_nose = 1
        self.num_heart = num_heart
    def eat(self):
        print("I can eat")
    def work(self):
        print("I can work")

class Male:
    def __init__(self, name):
        print("calling male init function")

        # Attribute belonging to the Male part of the object
        self.name = name

    def flirt(self):
        print("I can flirt")
    def work(self):
        print("I can code")

class Boy(Human, Male):
    # Boy inherits from both Human and Male
    # MRO: Boy → Human → Male → object
    def __init__(self, name, heart, language):
        print("calling boy init function")
        # Parent __init__ methods are NOT called automatically
        # because Boy has its own __init__ method.

        # Manually calling Human's __init__
        # This initializes num_eyes, num_nose and num_heart.
        Human.__init__(self, heart)

        # Manually calling Male's __init__
        # This initializes the name attribute.
        Male.__init__(self, name)

        # Initialize Boy's own attribute
        self.language = language

    def sleep(self):
        print("I can sleep")

    def work(self):
        # Boy has its own work(), so this takes priority.
        print("I can test")

    def display(self):
        print(f"Hi, I'm {self.name} and I work on {self.language}")

# Creating a Boy object
# if i dont define my own constructor then the order in which we specify the parent will called first 
#so here boy has its own constructor so Python automatically calls Boy.__init__()
boy_1 = Boy("Yeshu", 1, "Python")

# Attributes initialized by Human.__init__()
print(boy_1.num_nose)
print(boy_1.num_heart)
print(boy_1.num_eyes)

# Attribute initialized by Male.__init__()
print(boy_1.name)

# Attribute initialized by Boy.__init__()
print(boy_1.language)

# Calling Boy's method
boy_1.display()
````
