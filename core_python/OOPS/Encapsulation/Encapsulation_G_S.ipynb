

## Encapsulation

- Encapsulation means **bundling data (attributes) and methods together inside a class**.
- A normal class already provides this basic bundling.
- The main purpose of encapsulation is **controlling access to internal data/state**.
- It is commonly associated with **data hiding**.
- Private attributes, getters, setters, and methods can be used to control how internal data is accessed or modified.

Example:

    class BankAccount:
        def __init__(self):
            self.__balance = 1000

        def deposit(self, amount):
            if amount > 0:
                self.__balance += amount

        def get_balance(self):
            return self.__balance

- `__balance` → internal/private data
- `deposit()` → controls modification
- `get_balance()` → controls access

### Getter and Setter

- **Getter** → accesses/reads internal data.
- **Setter** → sets/modifies internal data.

## Abstraction

- Abstraction means **showing essential functionality while hiding unnecessary implementation details**.
- It focuses on **WHAT the object does**, rather than **HOW it does it**.

Example:

    account.deposit(500)

- The user knows that `deposit()` deposits money.
- The user does not need to know the internal steps used to update the balance.

## Abstract Class and Abstract Method

- **Abstract class** → a blueprint/contract for child classes.
- **Abstract method** → a method that concrete child classes must implement
## Main Difference

- **Abstraction** → decides **WHAT should be exposed** to the user and hides unnecessary complexity.
- **Encapsulation** → decides **HOW internal data/state is organized and controlled**.

```text
Abstraction
    ↓
WHAT should the user see/use?
    ↓
Essential functionality


Encapsulation
    ↓
How is internal data/state controlled?
    ↓
Data + methods + controlled access
```
##### Design vs Implementation
- Abstraction → Design level
- Decides what functionality should be exposed.
- Example: every payment should provide pay().
##### Encapsulation → Implementation level
- Organizes and controls the internal data and methods.
- Example: keeping __balance inside the object and modifying it through methods.

#### Relationship
- They are related but not the same.
- Abstraction focuses on hiding unnecessary complexity.
- Encapsulation focuses on bundling data and methods and controlling access to internal data.
- Therefore, encapsulation can be used to implement/support abstraction.

````python
class BankAccount:

    def __init__(self, name, balance):
        self.name = name              # Public attribute
        self.__balance = balance      # Private attribute
        self._account_type = "Savings"  # Protected attribute


    # Getter
    def get_balance(self):
        return self.__balance


    # Setter
    def set_balance(self, amount):
        if amount >= 0:
            self.__balance = amount
        else:
            print("Balance cannot be negative")


    # Method to deposit money
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"₹{amount} deposited")
        else:
            print("Invalid deposit amount")


    # Method to withdraw money
    def withdraw(self, amount):
        if amount <= 0:
            print("Invalid withdrawal amount")

        elif amount > self.__balance:
            print("Insufficient balance")

        else:
            self.__balance -= amount
            print(f"₹{amount} withdrawn")


    # Method to display account details
    def show_details(self):
        print(f"Name: {self.name}")
        print(f"Account Type: {self._account_type}")
        print(f"Balance: ₹{self.__balance}")
````

````python
account = BankAccount("Yeshu", 10000)
````

````python
print(account.name) #name is public, so it can be accessed directly.
````

````python
print(account._account_type) # Protected attribute can be accessed, but it's not recommended to do so outside the class.
````

````python
print(account.__balance) ## This will raise an AttributeError because __balance is private and cannot be accessed directly.
````

````python
print(account.get_balance()) # This will work because we are using the getter method to access the private attribute.
````

````python
account.set_balance(5000)
````

````python
account.deposit(2000)
````
