#### RAM, ROM, Disk & File Handling — Basics

##### RAM

- **RAM (Random Access Memory)** is temporary working memory.
- It is **hardware**.
- It stores data and instructions that running programs currently need.
- RAM is **volatile** → its contents are lost when power is turned off.

```text
Program running
      ↓
     RAM
      ↓
CPU works with the data
```

Example:

```python
x = 100
```

While the program is running, Python's objects and data are held in memory (RAM).

---

##### ROM

- **ROM (Read-Only Memory)** is a type of persistent memory traditionally used to store firmware.
- It is **hardware**.
- Firmware contains low-level instructions needed to initialize/start the computer.
- Modern computers commonly use **flash memory** for firmware, which can be updated.

```text
Computer ON
    ↓
Firmware
    ↓
Hardware initialization
    ↓
Operating System
```

---

##### Disk / SSD / HDD

- Disk refers to **persistent storage**.
- SSD and HDD are **hardware**.
- Data remains stored even after the computer is turned off.

Examples:

```text
Python files
Text files
Photos
Videos
Applications
Documents
```

##### SSD

- Uses flash memory.
- No moving mechanical parts.
- Generally much faster than HDD.

##### HDD

- Uses magnetic storage.
- Contains mechanically rotating disks and a read/write head.
- Generally slower than SSD.

---

##### RAM vs Disk

| RAM | Disk / SSD |
|---|---|
| Temporary | Permanent |
| Volatile | Non-volatile |
| Faster | Slower than RAM |
| Smaller | Usually much larger |
| Used by running programs | Used for long-term storage |

Simple flow:

```text
SSD
 ↓
Operating System loads data
 ↓
RAM
 ↓
CPU processes it
```

---

#### File Handling

- **File handling** means creating, reading, writing, updating, and managing files using Python.
- Files are normally stored on **disk/SSD**, while the data being actively processed is brought into **RAM**.

```text
Python program
      ↓
Operating System
      ↓
File System
      ↓
SSD / HDD
```

##### Basic File Operations

| Operation | Python |
|---|---|
| Open | `open()` |
| Read | `read()` |
| Read one line | `readline()` |
| Read all lines | `readlines()` |
| Write | `write()` |
| Move position | `seek()` |
| Get position | `tell()` |
| Close | `close()` |

##### File Modes

```text
"r" → read
"w" → write / overwrite
"a" → append
"x" → create new file
"rb" → read binary
"wb" → write binary
```

##### Recommended Way

```python
with open("data.txt", "r") as file:
    data = file.read()
```

- `with` automatically closes the file after the block finishes.

#### File Handling in Python

##### 1. Opening a File

```python
file = open("data.txt", "mode")
```

##### Common Modes

| Mode | Meaning | File exists? | File doesn't exist? |
|---|---|---|---|
| `"r"` | Read | Opens | ❌ `FileNotFoundError` |
| `"w"` | Write | Overwrites | ✅ Creates |
| `"a"` | Append | Adds at end | ✅ Creates |
| `"x"` | Create | ❌ `FileExistsError` | ✅ Creates |
| `"r+"` | Read + Write | Opens | ❌ `FileNotFoundError` |
| `"w+"` | Write + Read | Overwrites | ✅ Creates |
| `"a+"` | Append + Read | Opens | ✅ Creates |
| `"rb"` | Read binary | Opens | ❌ Error if missing |
| `"wb"` | Write binary | Overwrites | ✅ Creates |
| `"ab"` | Append binary | Adds at end | ✅ Creates |

---

#### 2. Important Difference Between `w`, `r`, `a`, and `x`

##### `"w"` — Write

```python
open("data.txt", "w")
```

- Creates the file if it doesn't exist.
- If it already exists, **old content is deleted/overwritten**.
- File pointer starts at the beginning.

##### `"r"` — Read

```python
open("data.txt", "r")
```

- Opens an existing file for reading.
- Does **not** create a missing file.
- File pointer starts at the beginning.

##### `"a"` — Append

```python
open("data.txt", "a")
```

- Creates the file if it doesn't exist.
- If it exists, new data is written at the **end**.
- Existing content is preserved.

##### `"x"` — Create

```python
open("data.txt", "x")
```

- Creates a **new file**.
- If the file already exists → `FileExistsError`.

---

### 3. `+` Modes

`+` means:

> **Allow both reading and writing.**

##### `"r+"`

```python
open("data.txt", "r+")
```

- Read + Write.
- File must already exist.
- Starts at the beginning.
- Does **not** automatically erase existing content.

##### `"w+"`

```python
open("data.txt", "w+")
```

- Write + Read.
- Creates the file if missing.
- If it exists → **erases existing content first**.
- Starts at the beginning.

##### `"a+"`

```python
open("data.txt", "a+")
```

- Append + Read.
- Creates the file if missing.
- Writing happens at the **end**.
- Existing content is preserved.
- The initial file position is at the end.

---

#### 4. Read and Write Order

For modes such as `"r+"`, `"w+"`, and `"a+"`, both operations are possible.

However, remember that the file has a **file pointer/position**.

```python
file.tell()
```

returns the current position.

```python
file.seek(position)
```

moves the position.

##### Example

```python
with open("data.txt", "w+") as file:
    file.write("Hello")
    file.seek(0)
    data = file.read()
    print(data)
```

Output:

```text
Hello
```

Why?

```text
write("Hello")
      ↓
pointer is at 5
      ↓
seek(0)
      ↓
pointer moves to beginning
      ↓
read()
      ↓
"Hello"
```

---

##### 5. Writing First, Then Reading

If you write first:

```python
file.write("Hello")
```

the pointer moves forward.

So:

```python
file.read()
```

may return nothing because the pointer is already at the end.

Use:

```python
file.seek(0)
```

before reading.

```python
with open("data.txt", "w+") as file:
    file.write("Hello")
    file.seek(0)
    print(file.read())
```

---

##### 6. Reading First, Then Writing

Example:

```python
with open("data.txt", "r+") as file:
    data = file.read()
    file.write("Hello")
```

After `read()`, the pointer is at the end of the existing content.

Therefore, `write()` starts writing from that position.

If you want to write somewhere else:

```python
file.seek(0)
file.write("Hello")
```

---

##### 7. `read()`

```python
data = file.read()
```

- Reads from the **current file position** to the end.
- Returns the content as a string in text mode.

```python
file.read(5)
```

- Reads up to 5 characters/bytes depending on the mode.

---

##### 8. `write()`

```python
file.write("Hello")
```

- Writes data at the **current file position**.
- Returns the number of characters written in text mode.

```python
n = file.write("Hello")
print(n)
```

Output:

```text
5
```

---

##### 9. `tell()`

```python
file.tell()
```

- Returns the **current file position**.

Example:

```python
with open("data.txt", "w+") as file:
    file.write("Hello")
    print(file.tell())
```

Output:

```text
5
```

---

##### 10. `seek()`

```python
file.seek(position)
```

- Moves the file pointer to a specified position.

Example:

```python
with open("data.txt", "r") as file:
    file.seek(2)
    print(file.read())
```

If the file contains:

```text
Hello
```

Output:

```text
llo
```

Because:

```text
H e l l o
0 1 2 3 4
    ↑
  seek(2)
```

---

##### 11. Text vs Binary

##### Text

```python
open("data.txt", "r")
```

Works with text such as:

```text
Hello
Python
```

Python handles the conversion between characters and bytes using an encoding.

##### Binary

```python
open("image.jpg", "rb")
```

Used for binary data such as:

```text
Images
Videos
PDFs
Audio
Executable files
```

Binary modes return/work with `bytes`.

---

##### 12. Recommended Pattern

Always prefer:

```python
with open("data.txt", "r") as file:
    data = file.read()
```

instead of:

```python
file = open("data.txt", "r")
data = file.read()
file.close()
```

`with` automatically closes the file when the block ends.

#### Quick Memory Table

```text
r   → read
w   → write / overwrite
a   → append
x   → create only

r+  → read + write, existing file required
w+  → write #+ read, creates/overwrites
a+  → append + read, creates if missing

rb  → read binary
wb  → write binary
ab  → append binary

read()  → read from current position
write() → write at current position
tell()  → show current position
seek()  → move current position

+       → allows both reading and writing
b       → binary mode
```

#### Most important concept

> **`read()` and `write()` work relative to the current file pointer. `tell()` tells you where the pointer is, and `seek()` moves it.**
