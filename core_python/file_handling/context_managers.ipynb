# Context Managers in Python

### 1. What is a Context Manager?
Imagine opening a file, working with it, and then closing it. What if an error occurs before you close it?

A context manager solves this problem by managing the setup and cleanup of a resource automatically.

### 2. Why do we need it?
Without a context manager, we must remember to close resources manually. If an exception interrupts execution, cleanup might be skipped.

### 3. Using `try-finally`
We can ensure cleanup using `finally`.

    file = open("data.txt", "w")
    try:
        file.write("Hello")
    finally:
        file.close()

Even if an exception occurs, `finally` executes and closes the file.

### 4. Using `with`
Python provides a simpler way to manage resources.

    with open("data.txt", "w") as file:
        file.write("Hello")

- `with` starts the managed block.
- `as file` stores the opened file object.
- When we leave the block, the file is automatically closed, even if an exception occurs.

### 5. Creating a Custom Context Manager Using a Class

We can create our own context manager using two special methods:

- `__enter__()` – prepares the resource when entering the `with` block and returns the object we want to use.
- `__exit__()` – runs when leaving the block, usually to perform cleanup.

    class FileManager:
        def __init__(self, filename):
            self.filename = filename

        def __enter__(self):
            self.file = open(self.filename, "w")
            return self.file

        def __exit__(self, exc_type, exc_value, traceback):
            self.file.close()

### 6. Understanding `__exit__()` Parameters

When an exception occurs inside the `with` block, Python provides:

- `exc_type` – type of exception.
- `exc_value` – exception object containing the error details.
- `traceback` – information about where the exception occurred.

If no exception occurs, all three are `None`.

Returning `True` suppresses the exception; returning `False` or `None` allows it to continue.

### 7. Using `contextlib.contextmanager`

Instead of creating a class, we can create a context manager using a generator function.

    from contextlib import contextmanager

    @contextmanager
    def open_file(filename):
        file = open(filename, "w")
        try:
            yield file
        finally:
            file.close()

- `@contextmanager` converts the generator function into a context manager.
- Code before `yield` performs setup.
- `yield file` provides the file object to the `with` block and pauses execution.
- After the block finishes, execution resumes after `yield`.
- `finally` performs cleanup.

### Quick Revision

| Concept | Purpose |
|---|---|
| Context Manager | Manages setup and cleanup |
| `with` | Enters and exits a managed block |
| `__enter__()` | Setup |
| `__exit__()` | Cleanup |
| `@contextmanager` | Creates a context manager using a generator |
| `yield` | Hands over control to the `with` block and pauses execution |

**Remember:** Context managers are useful for files, database connections, locks and other resources that need proper cleanup.

```python
file=open("file.txt","r")
try:
    file.write("hello everyone")
finally:
    file.close()
```

```python
with open("file.txt","r") as file:
    file.write("hello everyone")
```

```python
class File:

    def __init__(self, filename, mode):
        self.filename = filename
        self.mode = mode

    def __enter__(self):
        self.file = open(self.filename, self.mode)
        return self.file

    def __exit__(self, exc_type, exc_value, traceback):
        self.file.close()

with File("file.txt","w") as f:
    f.write("I'm Yeshwitha B M")
```
