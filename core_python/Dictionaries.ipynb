#### Dictionaries

- A dictionary is a built-in Python **data structure** that stores data in **key-value pairs**.
- Syntax: `dict_name = {key: value}`
- **Keys must be immutable/hashable** (e.g., `str`, `int`, `tuple`); values can be any valid Python object.
- Dictionaries are **mutable**.
- **Keys cannot be duplicated**; assigning the same key again replaces its previous value.
- We access values using their **keys**: `dict_name[key]`.
- Dictionaries preserve **insertion order** since Python 3.7.
- Before Python 3.7, dictionary order was **not guaranteed**.

```python
phone_no={
        'ram':1234,
        'shyam':3456,
        'mohan':8976,      
}
print(phone_no)
```

```python
#accessing
print(phone_no['mohan'])
```

```python
phone_no2={
        (1,2,3):1234,
        'shyam':3456,
        'mohan':8976, 
        (1,2,3):9876     
}
print(phone_no2[(1,2,3)])
print(phone_no2)
```

```python
phone_no3=dict({
        'ram':1234,
        'shyam':3456,
        'mohan':8976,} )     

print(phone_no3)
```

```python
phone_no4 = dict([
    ("Ram", 1234),
    ("Shyam", 3456),
    ("Mohan", 8976)
])

print(phone_no4)
```

```python
# Change value 
phone_no['Mohan']=9353
print(phone_no)
```

```python
#adding one more item 
phone_no['yeshu']=[111,22,333]
print(phone_no)
```

```python
# dictionary inside dictionary (nested dictionary)
phone_no['Ram']={'office':8766676,'home':878543}
print(phone_no)
print(phone_no['Ram'])
print(phone_no['Ram']['office'])
```

```python
data={
    1:'yeshu',
    2:'some',
    3:'mohan',
    0:'new',
    4:'lol',
    5:'what',
    6:'hello'
}
print(data[0])
```

```python
'''
d["name"] direct access → KeyError if missing
d.get("name") safe access → None if missing
'''
print(data.get(10))
```

```python
#delete any item 
del data[2]
c=data.pop(0)
print(data)
print(c)
```

```python
print(data.popitem()) # deletes last pair
print(data)
```

```python
#print(data.clear()) clears all item and left with empty dict
#print(data)
```

```python
# accessing keys and values we have built in method 
print(data)
print(data.keys())
print(data.values())
print(data.items()) #access both
```

```python
for i in data:
    print(i) #only keys 
    print(data[i])
```

```python
for i in data.keys():
    print(i) #only keys 
```

```python
for i in data.values():
    print(i) #only values
```

```python
for i in data.items():
    print(i) #both keys and values
```

```python
for i,j in data.items():
    print(i) #only key
    print(j)# only value
```

```python
# if we want to copy dict
print(data)
data2=data.copy()
print(data2)
```
