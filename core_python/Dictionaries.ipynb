#### Dictionaries

- A dictionary is a built-in Python **data structure** that stores data in **key-value pairs**.
- Syntax: `dict_name = {key: value}`
- **Keys must be immutable/hashable** (e.g., `str`, `int`, `tuple`); values can be any valid Python object.
- Dictionaries are **mutable**.
- **Keys cannot be duplicated**; assigning the same key again replaces its previous value.
- We access values using their **keys**: `dict_name[key]`.
- Dictionaries preserve **insertion order** since Python 3.7.
- Before Python 3.7, dictionary order was **not guaranteed**.

```python
phone_no={
        'ram':1234,
        'shyam':3456,
        'mohan':8976,      
}
print(phone_no)
```

```python
#accessing
print(phone_no['mohan'])
```

```python
phone_no2={
        (1,2,3):1234,
        'shyam':3456,
        'mohan':8976, 
        (1,2,3):9876     
}
print(phone_no2[(1,2,3)])
print(phone_no2)
```

```python
# Change value 
phone_no['Mohan']=9353
print(phone_no)
```

```python
#adding one more item 
phone_no['yeshu']=[111,22,333]
print(phone_no)
```

```python
student_marks={
    'yeshu':93,
    'harry':78,
    'dimpy':56,
    'rahul':41,
    'aniket':99,
    'prem':34
}
student_grade={}
for k,v in student_marks.items():
    if v >90:
        student_grade[k]='A+'
    elif v>80:
        student_grade[k]='A'
    elif v>70:
        student_grade[k]='B+'
    elif v>60:
        student_grade[k]='B'
    elif v>50:
        student_grade[k]='C'
    elif v>40:
        student_grade[k]='D'
    else:
        student_grade[k]='f'
print(student_grade)
```

#### Nested Dictionaries

- A dictionary inside another dictionary is called a **nested dictionary**.
- Multiple levels of nested dictionaries are also possible.

```python
student_data={
    'ram':{'roll_no':10,"age":20,"course":'python'},
    'mohan':{'roll_no':20,"age":22,"course":'java'}
}
print(student_data['mohan'])
print(student_data['mohan']['roll_no']) # accessing in nested dictionaries 
student_data['mohan']['phone_no']=987654 # adding new data inside a nested dict
print(student_data['mohan']) 
```

```python
student_data['phone_no'] # we dont have key 'phone_no' inside student data but have inside key of student_data
```

```python
#deleting inside nested dict
del student_data['mohan']['phone_no']
print(student_data['mohan'])
```

```python
# using pop 
student_data['ram'].pop('age')
```

```python
print(student_data)
```

```python
student_data=[
    {
        'name':'ram',
        'roll_no':10,
        "age":20,
        "course":'python'
    },
    {
        'name':'mohan',
        'roll_no':20,
        "age":22,
        "course":'java',
        'phone_no':[2837363,897765]
    }
]
def add_details(name,rolno,age,course):
    new_data={}
    new_data['name']=name
    new_data['rollno']=rolno
    new_data['age']=age
    new_data['course']=course
    student_data.append(new_data)

def add_new_data(**details):
    student_data.append(details)

add_details(name='yeshu',rolno=13,age=22,course='JS')    
add_new_data(name='shyam',roll_no=22,age=18,course="C++",phone_no=9327993827)
print(student_data)
```
